 ## 4. Combining DataFrames with Pandas<br>
 [See Notebook](./S5_Data_Cleaning/dsc-combining-dataframes-pandas-onl01-dtsc-ft-030220/index.ipynb)<br>
 When thinking about joins, it is easy to conceptualize them as Venn diagrams.<br>
 * An <b>Outer Join</b> returns all records from both tables<br>
 * An <b>Inner Join</b> returns only the records with matching keys in both tables<br>
 * A <b>Left Join</b> returns all the records from the left table, as well as any records from the right table that have a matching key with a record from the left table<br>
 * A <b>Right Join</b> returns all the records from the right table, as well as any records from the left table that have a matching key with a record from the right table<br>
