# Section 17 — Bayesian Statistics

The Bayesian framework treats probability as a degree of belief updated with evidence. It is increasingly relevant in 2026 — both in classical stats and as the theoretical backbone of LLM probabilistic reasoning.

> **2026 context — PyMC for Bayesian computation:** This section covers the mathematical foundations (Bayes' theorem, MLE, MAP). For practical Bayesian modelling with real posteriors, the modern tool is **PyMC** (formerly PyMC3, now v5). It lets you define probabilistic models in Python and sample posteriors via MCMC or variational inference.
>
> ```python
> # pip install pymc
> import pymc as pm
> import numpy as np
>
> data = np.array([2, 3, 3, 4, 3, 2, 3, 4])
>
> with pm.Model() as model:
>     mu = pm.Normal('mu', mu=0, sigma=10)        # prior
>     sigma = pm.HalfNormal('sigma', sigma=1)
>     obs = pm.Normal('obs', mu=mu, sigma=sigma, observed=data)  # likelihood
>     trace = pm.sample(1000, return_inferencedata=True)          # posterior
>
> pm.plot_posterior(trace)
> ```
