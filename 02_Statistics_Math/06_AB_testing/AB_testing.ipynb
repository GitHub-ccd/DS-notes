# Section 16 — A/B Testing

A/B testing is the primary method for causal inference in product and web analytics. The statistical methodology here is timeless; what has changed is the tooling around it.

> **2026 context:** In industry, A/B tests are rarely run with manual scipy code. Platforms like **Statsig**, **LaunchDarkly**, **Amplitude Experiment**, and **Optimizely** handle randomisation, exposure logging, and sequential testing automatically. Understanding the underlying statistics (this section) is what lets you interpret and challenge their outputs.
