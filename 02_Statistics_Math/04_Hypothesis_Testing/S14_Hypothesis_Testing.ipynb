## 5. Causal Inference — Awareness Note (2026)

Hypothesis testing tells you **whether** an effect exists. It does not tell you **why**, or what would happen if you intervened. Causal inference is the field that answers those questions.

Key concepts to be aware of:
| Concept | What it addresses |
|---------|-------------------|
| **Potential Outcomes / Rubin Causal Model** | Counterfactual reasoning — what *would have* happened |
| **DAGs (Directed Acyclic Graphs)** | Mapping causal structure, identifying confounders |
| **Difference-in-Differences** | Estimating treatment effects from observational data |
| **Instrumental Variables** | Handling unmeasured confounders |
| **RCTs vs Observational Studies** | Gold standard vs real-world constraints |
