## 5. Model Persistence
- [Pickle](./17_pickle.ipynb) — see security note inside

> **2026 note:** For sklearn models, prefer **joblib** over raw pickle — it's faster for large numpy arrays and is sklearn's own recommendation. `joblib.dump(model, 'model.joblib')` / `joblib.load('model.joblib')`. For production, consider **ONNX** (framework-agnostic) or **MLflow** model registry.
