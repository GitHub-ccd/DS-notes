## Understanding Mathematical Notation

When learning Data Science, running into dense, arcane-looking mathematical equations is an inevitability. For instance, here's an equation for a machine learning algorithm called a _Support Vector Machine_ that you'll learn about later in the course:

$ \LARGE \text{minimize:} $

$$ \LARGE W(\alpha) = - \sum_{i=1}^{\ell}\alpha_i + \frac{1}{2}\sum_{i=1}^{\ell}\sum_{j=1}^{\ell}y_iy_j\alpha_i\alpha_j\textbf{x}_i\textbf{x}_j$$

$ \LARGE \text{subject to:} $ $$ \LARGE \sum_{\substack{i=1 \\ 0\leq\alpha_i\leq C}}^{\ell}y_i \alpha_i = 0 $$

<strong><em>Don't stare at this too long...</em></strong>

At first glance, equations like this often seem impenetrable. If seeing math like this makes you nervous, don't worry — you're not alone! This is a particularly dense example, but at the end of the day it's just a recipe for an algorithm. Whereas we express algorithms in code, mathematicians prefer **_Mathematical Notation_** — big ideas in small, dense packages full of Greek letters, subscripts, and superscripts.

The first thing to remember: **do not try to memorize every symbol**. Instead, keep a cheat sheet handy and use Google freely. There is no shame in looking it up — senior data scientists do it constantly.

Two types of notation are worth examining in depth: **Summations** and **Products**.

---

## Sigma Notation (Summations)

In the equation above, you'll notice:

$$\LARGE \sum $$

This is called **Sigma Notation** and represents **Summation**.

Consider the following code:

```python
sum_total = 0
n = 5
for i in range(1, n+1):
    sum_total += i
```

A mathematician would express the same algorithm as:

$$\LARGE \sum_{n=1}^{5} x $$

The equation says: start at $n=1$, stop at $n=5$ (inclusive), and sum each value $x$. Written out explicitly:

$$\LARGE \sum_{n=1}^{5} x = 1 + 2 + 3 + 4 + 5 $$

You can put any expression to the right of the Sigma symbol. For example:

```python
def f(x):
    return x**2 - 2

sum_total = sum(f(i) for i in range(1, 6))
```

In Sigma Notation:

$$\LARGE \sum_{n=1}^{5}x^2-2 = (1^2 - 2) + (2^2 - 2) + (3^2 - 2) + (4^2 - 2) + (5^2 - 2) $$

When working with arrays, subscripts indicate indexing:

```python
w = [2.2, 3, -2, 4]
x = [0.25, 0, 1, -2]
sum_total = sum(w[i] * x[i] for i in range(len(w)))
```

$$\LARGE \sum_{i=1}^{n} w_i x_i $$

> **Note:** Sigma notation starts counting at 1; Python starts at 0. They refer to the same thing — the first element — but the indexing convention differs. Keep this in mind when translating formulas to code.

## Pi Notation (Products)

**Pi Notation** works similarly to Sigma Notation, but it **multiplies** each step instead of adding:

```python
total = 1
for i in range(1, 6):
    total *= i + 2
```

In Pi Notation:

$$ \LARGE \prod_{i=1}^{5} i + 2 = (1 + 2) \times (2 + 2) \times (3 + 2) \times (4 + 2) \times (5 + 2)$$

The only difference from Sigma: $\sum$ adds each step, $\prod$ multiplies.

## Common Greek Letters in Data Science

| Symbol | Name | Common Use |
|--------|------|------------|
| $\mu$ | mu | Population mean |
| $\sigma$ | sigma (lowercase) | Standard deviation |
| $\sigma^2$ | sigma-squared | Variance |
| $\Sigma$ | Sigma (uppercase) | Summation |
| $\Pi$ | Pi (uppercase) | Product |
| $\alpha$ | alpha | Learning rate, significance level |
| $\beta$ | beta | Regression coefficients |
| $\theta$ | theta | Model parameters |
| $\epsilon$ | epsilon | Error term |
| $\lambda$ | lambda | Regularization parameter |
| $\rho$ | rho | Correlation coefficient |
