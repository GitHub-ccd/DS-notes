## Part 1 — Implementing the Functions

```python
def get_mean(data):
    """
    Calculate the mean: sum of all values / count of values.
    Formula: x̄ = (1/n) * Σx_i
    """
    mean = sum(data) / len(data)
    return round(mean, 2)

# Test
print(get_mean([5, 4, 1, 3, 2]))  # Expected: 3.0
print(get_mean([4, 2, 3, 1]))     # Expected: 2.5
```

```python
def get_median(data):
    """
    Sort the data, then pick the middle value.
    For even n, average the two middle values.
    """
    data_sorted = sorted(data)
    n = len(data_sorted)
    mid = n // 2
    if n % 2 == 1:
        return data_sorted[mid]
    else:
        return (data_sorted[mid - 1] + data_sorted[mid]) / 2

# Test
print(get_median([5, 4, 1, 3, 2]))  # Expected: 3
print(get_median([4, 2, 3, 1]))     # Expected: 2.5
```

```python
def get_mode(data):
    """
    Build a frequency dict, return all values tied for max frequency.
    """
    freq = {}
    for val in data:
        freq[val] = freq.get(val, 0) + 1
    max_freq = max(freq.values())
    return [k for k, v in freq.items() if v == max_freq]

# Test
print(get_mode([1, 2, 3, 5, 5, 4]))       # Expected: [5]
print(get_mode([1, 1, 1, 2, 3, 4, 5, 5, 5]))  # Expected: [1, 5]
```

```python
def get_variance(sample):
    """
    Sample variance: Σ(x_i - x̄)² / (n-1)
    Divide by n-1 (Bessel's correction) for unbiased estimation.
    """
    sample_mean = get_mean(sample)
    sq_deviations = [(x - sample_mean) ** 2 for x in sample]
    return round(sum(sq_deviations) / (len(sample) - 1), 2)

# Test
print(get_variance([1, 2, 3, 5, 5, 4]))  # Expected: ~2.67
print(get_variance([1, 1, 1, 2, 3, 4, 5, 5, 5]))  # Expected: ~3.25
```

```python
from math import sqrt

def get_stddev(data):
    """
    Standard deviation: sqrt of sample variance.
    Returns result in the same units as the data.
    """
    return round(sqrt(get_variance(data)), 2)

# Test
print(get_stddev([120, 112, 131, 211, 312, 90]))  # Expected: 84.03
```

## Part 2 — Applying to the NHIS Height Data

```python
print("=== Raw height data ===")
print(f"Mean:     {get_mean(height)}")
print(f"Median:   {get_median(height)}")
print(f"Mode:     {get_mode(height)}")
print(f"Variance: {get_variance(height)}")
print(f"Std Dev:  {get_stddev(height)}")
# Mean ~69.58, Median ~67, Std Dev ~9.37
```

```python
def filter_list(data, threshold=80):
    """Remove values above threshold to handle obvious outliers."""
    return [x for x in data if x <= threshold]

filtered_height = filter_list(height)
print(f"Records after filter: {len(filtered_height)} (removed {len(height) - len(filtered_height)})")
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].hist(height, bins=8)
axes[0].set_title('Original (with outliers)')
axes[1].hist(filtered_height, bins=8)
axes[1].set_title('Filtered (≤ 80 inches)')
plt.show()
```

```python
print("=== Filtered height data ===")
print(f"Mean:     {get_mean(filtered_height)}")  # Expected: ~66.85
print(f"Median:   {get_median(filtered_height)}")

print("\n=== Comparison ===")
print(f"Mean shift due to outliers: {get_mean(height) - get_mean(filtered_height):.2f} inches")
print(f"Median shift: {get_median(height) - get_median(filtered_height)} — much smaller")
```

```python
# Final box plot
import numpy as np
plt.boxplot(filtered_height)
plt.title('Height Box Plot (filtered)')
plt.show()
```

```python
import numpy as np

print("Our implementation vs NumPy:")
print(f"  Mean:    {get_mean(height)} vs {np.mean(height):.2f}")
print(f"  Median:  {get_median(height)} vs {np.median(height)}")
print(f"  Std Dev: {get_stddev(height)} vs {np.std(height, ddof=1):.2f}")
# Results should match (or be within rounding)
```

| Function | What it measures |
|----------|------------------|
| `get_mean()` | Center of the data |
| `get_median()` | Middle value (robust to outliers) |
| `get_mode()` | Most frequent value |
| `get_variance()` | Average squared deviation from mean |
| `get_stddev()` | Spread in original units |

Key observations from the NHIS data:
