# Measures of Central Tendency

## Introduction

When working with large datasets, we prefer features that summarize data in a concise format rather than discussing individual values. In this lesson, we look at **measures of central tendency** — statistics that describe the typical or central value in a dataset.

## Objectives

You will be able to:

* Calculate the mean, median, and mode
* Explain when each measure is most appropriate
* Identify the effect of outliers and distribution shape on each measure

---

## The Three Measures

There are three main measures of central tendency: the **mean**, the **median**, and the **mode**.

---

## Mean

The **Mean** (Arithmetic Average) is the sum of all values divided by the number of data points:

$$ \Large\bar X = \dfrac{\sum X}{N} $$

Example — retirement ages of 11 individuals:
```
54, 54, 54, 55, 56, 57, 57, 58, 58, 60, 60
```
Mean = 623 / 11 = **56.6**

### Sample Mean vs. Population Mean

The **sample mean** ($\bar{x}$) is calculated from a subset of the population. The **population mean** ($\mu$, "mew") covers the entire population. The formula is the same, but the notation differs — and this distinction matters for hypothesis testing later in the course.

---

## Median

The **Median** is the middle value when data is sorted.

- **Odd number of values:** middle item. For our 11 retirement ages, the 6th value = **57**.
- **Even number of values:** average of the two middle items.

Example with 10 weights: `55, 56, 56, 58, 60, 61, 63, 64, 70, 78`  
Median = (60 + 61) / 2 = **60.5**

**When to use the median:** when outliers are present. If 9 people each earn \$50k and one hedge fund manager earns \$20M, the mean salary is over \$1.86M — technically accurate but meaningless as a typical value. The median stays at \$50k.

---

## Mode

The **Mode** is the value that appears most frequently.

In our retirement data, 54 appears 3 times — **mode = 54**.

A dataset can have no mode, one mode, or multiple modes (bimodal / multimodal). Mode is especially useful for **categorical data** and for **filling missing values**.

---

## Visualizing Central Tendency

````python
import matplotlib.pyplot as plt
%matplotlib inline

x = [54, 54, 54, 55, 56, 57, 57, 58, 58, 60, 60]
plt.hist(x, bins=5)
plt.axvline(sum(x)/len(x), color='red', linestyle='--', label=f'Mean = {sum(x)/len(x):.1f}')
plt.axvline(sorted(x)[len(x)//2], color='green', linestyle='--', label='Median = 57')
plt.axvline(54, color='orange', linestyle='--', label='Mode = 54')
plt.title('Retirement Ages')
plt.legend()
plt.show()
````

## Distribution Shape and Central Tendency

### Symmetrical Distributions

For symmetric distributions, mean, median, and mode coincide at the center.

<img src="assets/measures_of_central_tendency/image_sym.png" width="450">

### Skewed Distributions

In skewed distributions, the mean is pulled toward the tail. The median is more robust and is often preferred.

<img src="assets/measures_of_central_tendency/image_pos.png" width="450">

<img src="assets/measures_of_central_tendency/image_neg.png" width="450">

### Outliers

Outliers affect the mean strongly but barely affect the median or mode:

Original: `54, 54, 54, 55, 56, 57, 57, 58, 58, 60, 60` → Mean = 56.6, Median = 57  
With outlier: `54, 54, 54, 55, 56, 57, 57, 58, 58, 60, **81**` → Mean = 58.5, Median = **57** (unchanged)

## Calculating with Python

````python
import statistics
import numpy as np

data = [54, 54, 54, 55, 56, 57, 57, 58, 58, 60, 60]

print(f"Mean:   {np.mean(data):.2f}")
print(f"Median: {np.median(data)}")
print(f"Mode:   {statistics.mode(data)}")
````

````python
# Q5: Which measure is most sensitive to outliers — mean, median, or mode?
# Demonstrate with an example.
clean = [10, 12, 11, 13, 10, 12]
with_outlier = clean + [1000]

print("Without outlier:")
print(f"  Mean: {np.mean(clean):.2f}, Median: {np.median(clean)}")
print("With outlier (1000):")
print(f"  Mean: {np.mean(with_outlier):.2f}, Median: {np.median(with_outlier)}")
# Mean shifts dramatically; median barely changes
````

## Summary

In this lesson, you learned the three measures of central tendency and when to use each:

| Measure | Best for | Sensitive to outliers? |
|---------|----------|------------------------|
| Mean | Symmetric, normally distributed data | Yes |
| Median | Skewed data, data with outliers | No |
| Mode | Categorical data, frequency analysis | No |

These measures will appear throughout the curriculum — in hypothesis testing, regression, and beyond. The next lesson covers **measures of dispersion**, which describe how spread out data is around its center.
