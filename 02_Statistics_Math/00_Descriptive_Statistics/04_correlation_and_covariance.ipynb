# Covariance and Correlation

## Introduction

In this lesson, you'll learn how **covariance** and **correlation** extend variance from a single variable to the relationship between two variables. These measures are foundational to regression, dimensionality reduction, and most supervised ML algorithms.

## Objectives

You will be able to:

- Calculate covariance and correlation from scratch in Python
- Interpret the sign and magnitude of each measure
- Distinguish correlation from causation
- Use Pandas and NumPy to compute these measures efficiently

---

## Variance — a quick recap

Recall variance for a single variable:

$$\sigma^2 = \dfrac{1}{n}\displaystyle\sum_{i=1}^{n}(x_i-\mu)^2$$

---

## Covariance

**Covariance** measures how two variables change together:

$$\sigma_{XY} = \dfrac{1}{n}\displaystyle\sum_{i=1}^{n}(x_i -\mu_x)(y_i - \mu_y)$$

**Interpretation:**
- **Positive covariance** → both variables move in the same direction
- **Negative covariance** → variables move in opposite directions
- **Near zero** → no linear relationship

**Key limitation:** covariance values range from $-\infty$ to $+\infty$, so you cannot compare covariances across datasets with different scales.

**Example — ice cream vendors (January-May):**

| Month | Vendor A | Vendor B |
|-------|----------|----------|
| Jan | 100 | 220 |
| Feb | 200 | 400 |
| Mar | 300 | 600 |
| Apr | 500 | 900 |
| May | 900 | 1300 |

When A goes up, B goes up → positive covariance. But how strong is the relationship? We need **correlation** to answer that.

---

## Pearson Correlation Coefficient

**Correlation** standardizes covariance by the variability in each variable:

$$ r = \frac{\sum_{i=1}^{n}(x_i -\bar x)(y_i - \bar y)} {\sqrt{\sum_{i=1}^{n}(x_i - \bar x)^2 \sum_{i=1}^{n}(y_i-\bar y)^2}}$$

**Key properties:**
- Always between **-1 and +1**
- **+1** = perfect positive linear relationship
- **-1** = perfect negative linear relationship
- **0** = no linear relationship
- Dimensionless — can be compared across different datasets

## Correlation is Not Causation

> When A and B are correlated, it does not mean A causes B.

Classic example: **ice cream sales and homicide rates** in New York are positively correlated. The hidden factor is **weather** — hot weather independently increases both. No causal link between ice cream and crime.

<img src="assets/correlation_covariance/hidden_factor.png" width="700">

Another famous spurious correlation: **Nicolas Cage movie releases vs. pool drowning deaths**.

<img src="assets/correlation_covariance/cage.png" width="700">

**Key rule:** high correlation is necessary but not sufficient for causation. Always look for confounders.

## Building from Scratch — Lab

```python
# Dataset: 20 heights (inches) and weights (pounds)
height = [68, 71, 61, 69, 71, 58, 72, 73, 58, 74, 
          61, 59, 69, 68, 64, 69, 72, 66, 65, 69]
weight = [165, 201, 140, 170, 192, 125, 195, 205, 
          115, 210, 135, 125, 172, 175, 145, 170, 
          200, 155, 150, 171]
```

```python
def mean_normalize(var):
    """Subtract mean from each element — centers the variable at zero."""
    mean = sum(var) / len(var)
    return [x - mean for x in var]

print(mean_normalize([1, 2, 3, 4, 5]))
# Expected: [-2.0, -1.0, 0.0, 1.0, 2.0]
```

```python
def dot_product(x, y):
    """Element-wise multiply then sum — the numerator of both covariance and correlation."""
    return sum(xi * yi for xi, yi in zip(x, y))

print(dot_product([1, 2, 3], [4, 5, 6]))  # Expected: 32
```

```python
def covariance(var1, var2):
    if len(var1) != len(var2):
        raise ValueError('Variables must have the same length.')
    x = mean_normalize(var1)
    y = mean_normalize(var2)
    return dot_product(x, y) / (len(var1) - 1)

cov = covariance(height, weight)
print(f"Covariance (height, weight): {cov:.2f}")
# Expected: ~144.76
```

```python
import matplotlib.pyplot as plt
%matplotlib inline

plt.scatter(height, weight)
plt.xlabel('Height (in)')
plt.ylabel('Weight (lb)')
plt.title('Height vs Weight')
plt.show()
# Positive covariance is visible — taller people tend to weigh more
```

```python
import math

def correlation(var1, var2):
    if len(var1) != len(var2):
        raise ValueError('Variables must have the same length.')
    x = mean_normalize(var1)
    y = mean_normalize(var2)
    numerator = dot_product(x, y)
    denominator = math.sqrt(sum(xi**2 for xi in x) * sum(yi**2 for yi in y))
    return round(numerator / denominator, 2)

r = correlation(height, weight)
print(f"Pearson r (height, weight): {r}")
# Expected: 0.98 — very strong positive relationship
```

A correlation of 0.98 is extremely high. This sample of 20 people happens to correlate very strongly, but sample size matters — different samples from the same population can yield different correlations.

> A variable correlated with itself always has $r = 1.0$.

```python
# Verify with NumPy and Pandas
import numpy as np
import pandas as pd

df = pd.DataFrame({'height': height, 'weight': weight})
print("NumPy correlation matrix:")
print(np.corrcoef(height, weight))
print("\nPandas corr():")
print(df.corr())
```

## Summary

| Measure | Range | Interpretable? | Use |
|---------|-------|----------------|-----|
| Covariance | $(-\infty, +\infty)$ | No | Directional relationship only |
| Pearson $r$ | $[-1, 1]$ | Yes | Strength and direction of linear relationship |

Key takeaways:
- Covariance tells you the **direction** of a relationship but is hard to interpret in absolute terms.
- Correlation is the **standardized** version — always interpretable, always comparable.
- **Correlation ≠ causation** — always look for hidden factors and confounders.
