> **2026 context:** MLPs (fully connected networks) are covered here as the foundation. For images, **CNNs** supersede MLPs (see [Convolutional_Neural_Networks/](../Convolutional_Neural_Networks/Convolutional_Neural_Networks.ipynb)). For sequences and language, **Transformers** have replaced both RNNs and MLPs — see MOD_5. In PyTorch, the equivalent of a Keras `Sequential` model is `torch.nn.Sequential` or a custom `nn.Module` subclass.
