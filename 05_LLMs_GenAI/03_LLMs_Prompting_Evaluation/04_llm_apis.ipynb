# 04 — LLM APIs

The three dominant LLM API providers — Anthropic, OpenAI, and Google — all expose similar REST APIs but with different authentication, parameter names, and feature rollout schedules. This notebook covers the patterns that are common across providers.

## Anthropic API

```python
import anthropic

client = anthropic.Anthropic()  # reads ANTHROPIC_API_KEY from env

# Basic message
msg = client.messages.create(
    model='claude-haiku-4-5-20251001',
    max_tokens=256,
    system='You are a concise technical assistant.',
    messages=[
        {'role': 'user', 'content': 'What is the difference between precision and recall?'}
    ]
)
print(msg.content[0].text)
print(f'\nInput tokens: {msg.usage.input_tokens}, Output tokens: {msg.usage.output_tokens}')
```

```python
# Multi-turn conversation
history = []

def chat(user_message):
    history.append({'role': 'user', 'content': user_message})
    msg = client.messages.create(
        model='claude-haiku-4-5-20251001',
        max_tokens=256,
        messages=history,
    )
    reply = msg.content[0].text
    history.append({'role': 'assistant', 'content': reply})
    return reply

print(chat('Name three Python data science libraries.'))
print()
print(chat('Which one is best for deep learning?'))  # remembers context
```

## Streaming responses

```python
# Streaming: print tokens as they arrive — better UX for long responses
with client.messages.stream(
    model='claude-haiku-4-5-20251001',
    max_tokens=256,
    messages=[{'role': 'user', 'content': 'Write a short poem about neural networks.'}],
) as stream:
    for text in stream.text_stream:
        print(text, end='', flush=True)
print()  # newline after streaming
```

## Structured output — JSON extraction

```python
import json

def extract_json(text, schema_description, data):
    msg = client.messages.create(
        model='claude-haiku-4-5-20251001',
        max_tokens=512,
        system=f'You extract structured data. Respond ONLY with valid JSON matching this schema: {schema_description}. No explanation, no markdown.',
        messages=[{'role': 'user', 'content': data}],
    )
    return json.loads(msg.content[0].text)

schema = '{"products": [{"name": str, "price": float, "in_stock": bool}]}'
text = 'We have: Laptop at $999 (available), Mouse at $25 (out of stock), Monitor at $349 (available)'

result = extract_json(text, schema, text)
print(json.dumps(result, indent=2))
```

## OpenAI-compatible API

Most providers (OpenAI, Mistral, Together, Groq, Ollama) use the same `/v1/chat/completions` endpoint format.

```python
# pip install openai
from openai import OpenAI

# OpenAI
openai_client = OpenAI()  # reads OPENAI_API_KEY from env

# Mistral (or any OpenAI-compatible endpoint)
# mistral_client = OpenAI(
#     base_url='https://api.mistral.ai/v1',
#     api_key=os.environ['MISTRAL_API_KEY'],
# )

# Ollama (local — no API key needed)
# ollama_client = OpenAI(base_url='http://localhost:11434/v1', api_key='ollama')

# Call is identical regardless of provider
def openai_ask(client, model, user, system='You are a helpful assistant.'):
    response = client.chat.completions.create(
        model=model,
        messages=[
            {'role': 'system', 'content': system},
            {'role': 'user', 'content': user},
        ],
        max_tokens=256,
    )
    return response.choices[0].message.content

# result = openai_ask(openai_client, 'gpt-4o-mini', 'Explain embeddings in one sentence.')
# print(result)
print('OpenAI-compatible call defined — uncomment to run with API key')
```

## Prompt caching (Anthropic)

Prompt caching stores the KV cache of a prompt prefix server-side. Re-sending the same prefix costs ~10% of normal input token price and is ~5× faster.

```python
# Mark a large system prompt as cacheable
# Minimum 1024 tokens (Haiku) or 2048 tokens (Sonnet/Opus) to cache

long_document = """[Imagine this is a 5000-token document — a codebase, a legal contract, etc.]
""" * 50  # simulated long content

# Cache this expensive prefix
cached_response = client.messages.create(
    model='claude-haiku-4-5-20251001',
    max_tokens=256,
    system=[
        {
            'type': 'text',
            'text': long_document,
            'cache_control': {'type': 'ephemeral'},  # cache this block
        },
        {
            'type': 'text',
            'text': 'Answer questions about the document above.',
        }
    ],
    messages=[{'role': 'user', 'content': 'Summarize in one sentence.'}],
)
print('Cache tokens written:', cached_response.usage.cache_creation_input_tokens)
print('Cache tokens read:', cached_response.usage.cache_read_input_tokens)
```

## API comparison

| Feature | Anthropic | OpenAI | Google (Gemini) |
|---------|-----------|--------|----------------|
| SDK | `anthropic` | `openai` | `google-generativeai` |
| Message format | `system` param + `messages` | `messages` with `role=system` | `system_instruction` + `contents` |
| Structured output | JSON via prompt | `response_format` | `response_mime_type` |
| Streaming | `.stream()` context | `stream=True` | `stream=True` |
| Prompt caching | `cache_control` per block | Automatic (5-min TTL) | Context caching API |
| Token counting | `usage` in response | `usage` in response | `count_tokens()` pre-call |
