# 02 — Prompt Engineering

Prompt engineering is the practice of designing inputs to get reliable, high-quality outputs from LLMs. It's the highest-leverage skill for getting value out of LLMs before committing to fine-tuning.

## Basic patterns

```python
import anthropic  # pip install anthropic

client = anthropic.Anthropic()  # reads ANTHROPIC_API_KEY from env

def ask(system, user, model='claude-haiku-4-5-20251001', max_tokens=512):
    msg = client.messages.create(
        model=model,
        max_tokens=max_tokens,
        system=system,
        messages=[{'role': 'user', 'content': user}],
    )
    return msg.content[0].text

# Zero-shot
result = ask(
    system='You are a helpful assistant.',
    user='Classify this review as POSITIVE or NEGATIVE: "The battery lasts forever but the screen is dim."'
)
print('Zero-shot:', result)
```

```python
# Few-shot — show examples in the prompt
few_shot_prompt = """
Classify each review as POSITIVE, NEGATIVE, or MIXED.

Review: "Great sound quality and comfortable." → POSITIVE
Review: "Broke after two days. Terrible." → NEGATIVE
Review: "Fast delivery but wrong item sent." → MIXED
Review: "The battery lasts forever but the screen is dim." →"""

result = ask(system='You are a helpful assistant.', user=few_shot_prompt)
print('Few-shot:', result.strip())
```

## System prompt — set persona and constraints

```python
# System prompts define behaviour, tone, output format, and constraints
result = ask(
    system="""You are a senior Python engineer doing code review.
Be concise. Always:
1. Identify the main issue in one sentence
2. Show the fixed code
3. Explain WHY in one sentence""",
    user="""
def get_user(users, id):
    for user in users:
        if user['id'] == id:
            return user"""
)
print(result)
```

## Structured output — JSON mode

```python
import json

result = ask(
    system='You extract structured data. Always respond with valid JSON only. No explanation.',
    user="""
Extract from: 'John Smith, 34, joined Acme Corp in March 2021 as a Senior Engineer in London.'

Schema: {"name": str, "age": int, "company": str, "start_date": str, "role": str, "location": str}"""
)

data = json.loads(result)
print(json.dumps(data, indent=2))
```

## Temperature and sampling

```python
# temperature=0 → deterministic (greedy) — use for factual/structured tasks
# temperature=0.7–1.0 → creative variation — use for writing/brainstorming

for temp in [0.0, 0.7, 1.2]:
    msg = client.messages.create(
        model='claude-haiku-4-5-20251001',
        max_tokens=50,
        temperature=temp,
        messages=[{'role': 'user', 'content': 'Complete: The best way to learn ML is to'}],
    )
    print(f'temp={temp}: {msg.content[0].text[:80]}...')
```

## Common failure modes and fixes

| Problem | Symptom | Fix |
|---------|---------|-----|
| Vague output | Response is too general | Add format constraints: "respond in bullet points", "max 3 sentences" |
| Wrong format | JSON broken, markdown leaking | Add negative constraint: "respond ONLY with JSON, no other text" |
| Hallucination | Confident but wrong facts | Add: "If unsure, say 'I don't know'. Do not guess." |
| Instruction ignored | Model does its own thing | Use XML-style delimiters for content: `<document>...</document>` |
| Inconsistent | Different answers on same input | Lower temperature; add few-shot examples of desired output |
