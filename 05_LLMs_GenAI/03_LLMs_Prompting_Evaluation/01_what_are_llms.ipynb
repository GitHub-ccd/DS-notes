# 01 — What Are LLMs?

## From language model to LLM

A **language model** assigns a probability to a sequence of tokens. The simplest version: given tokens so far, predict the next one. This is **Causal Language Modelling (CLM)**.

A **Large Language Model** is a language model trained on enormous corpora (trillions of tokens) at scale — typically >1B parameters. Scale alone produces surprising capabilities.

## Three stages of training

### 1. Pre-training
- **Data:** web text (Common Crawl), books, code, Wikipedia, papers
- **Objective:** next-token prediction on trillions of tokens
- **Cost:** millions of dollars in compute
- **Result:** a base model that can complete text — but doesn't follow instructions

### 2. Supervised Fine-Tuning (SFT) / Instruction Tuning
- **Data:** thousands to millions of (instruction, response) pairs
- **Objective:** standard cross-entropy on the response tokens
- **Result:** a model that follows instructions reasonably well

### 3. Reinforcement Learning from Human Feedback (RLHF)
- **Data:** human rankings of model outputs
- **Objective:** train a reward model, then fine-tune with PPO (or DPO) to maximise reward
- **Result:** a model aligned with human preferences — less harmful, more helpful, more honest

## Scaling laws

**Chinchilla (Hoffmann et al., 2022):** for a given compute budget, train a *smaller* model on *more* tokens. Optimal: ~20 tokens per parameter. GPT-3 (175B params) was undertrained — Llama 3 8B trained on 15T tokens often matches it.

## Emergent capabilities

Capabilities that appear suddenly at scale with no prior evidence at smaller sizes: few-shot learning, chain-of-thought reasoning, arithmetic, code generation. Not fully understood — active research area.
