# 02 — Multi-Head Attention

## Why multiple heads?

A single attention head attends to the sequence with one set of Q/K/V projections. But language has multiple simultaneous relationships — syntax, coreference, semantic similarity, positional proximity. Different heads can specialise:

- Head 1: syntactic dependencies (subject → verb)
- Head 2: coreference ("it" → "the model")
- Head 3: local context (adjacent words)
- etc.

## How it works

Run $h$ attention heads in **parallel** on **lower-dimensional projections**, then **concatenate** and project:

$$\text{MultiHead}(Q, K, V) = \text{Concat}(\text{head}_1, ..., \text{head}_h) W^O$$

where $\text{head}_i = \text{Attention}(QW_i^Q, KW_i^K, VW_i^V)$

Each head uses dimension $d_k = d_{\text{model}} / h$, so total compute stays constant vs one full-dimensional head.

```python
import torch
import torch.nn as nn
import math
import torch.nn.functional as F

class MultiHeadAttention(nn.Module):
    def __init__(self, d_model, num_heads):
        super().__init__()
        assert d_model % num_heads == 0
        self.d_model = d_model
        self.num_heads = num_heads
        self.d_k = d_model // num_heads

        self.W_q = nn.Linear(d_model, d_model)
        self.W_k = nn.Linear(d_model, d_model)
        self.W_v = nn.Linear(d_model, d_model)
        self.W_o = nn.Linear(d_model, d_model)

    def split_heads(self, x):
        # (batch, seq, d_model) → (batch, heads, seq, d_k)
        B, S, _ = x.shape
        return x.view(B, S, self.num_heads, self.d_k).transpose(1, 2)

    def forward(self, x, mask=None):
        B, S, _ = x.shape
        Q = self.split_heads(self.W_q(x))
        K = self.split_heads(self.W_k(x))
        V = self.split_heads(self.W_v(x))

        scores = torch.matmul(Q, K.transpose(-2, -1)) / math.sqrt(self.d_k)
        if mask is not None:
            scores = scores.masked_fill(mask == 0, float('-inf'))
        weights = F.softmax(scores, dim=-1)

        context = torch.matmul(weights, V)  # (B, heads, S, d_k)
        context = context.transpose(1, 2).contiguous().view(B, S, self.d_model)  # concat
        return self.W_o(context), weights


# Test: 2 sequences of length 6, d_model=64, 8 heads
torch.manual_seed(0)
mha = MultiHeadAttention(d_model=64, num_heads=8)
x = torch.randn(2, 6, 64)  # batch=2, seq=6, d_model=64
out, attn_weights = mha(x)
print(f'Output shape:  {out.shape}')          # (2, 6, 64)
print(f'Weights shape: {attn_weights.shape}')  # (2, 8, 6, 6) — 8 heads
```

## PyTorch built-in

In practice you'd use `torch.nn.MultiheadAttention` rather than implementing from scratch:

```python
mha_builtin = nn.MultiheadAttention(embed_dim=64, num_heads=8, batch_first=True)
out_builtin, weights_builtin = mha_builtin(x, x, x)  # self-attention: Q=K=V=x
print(f'Built-in output: {out_builtin.shape}')  # (2, 6, 64)
```

## Typical hyperparameters

| Model | d_model | num_heads | d_k (per head) | Layers |
|-------|---------|-----------|----------------|--------|
| BERT-base | 768 | 12 | 64 | 12 |
| GPT-2 small | 768 | 12 | 64 | 12 |
| GPT-3 (175B) | 12288 | 96 | 128 | 96 |
| Llama 3 8B | 4096 | 32 | 128 | 32 |

**Key insight:** $d_k = d_{\text{model}} / h$ stays roughly constant as models scale — the number of *layers* and *heads* grows, not $d_k$.
