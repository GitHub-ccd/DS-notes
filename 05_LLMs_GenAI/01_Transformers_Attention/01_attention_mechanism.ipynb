# 01 — The Attention Mechanism

## The problem with RNNs

Before transformers, sequence modelling used RNNs (LSTMs, GRUs). They process tokens *sequentially*, passing a fixed-size hidden state forward. Two problems:

1. **Long-range dependencies:** by the time token 512 is processed, information from token 1 has been compressed into a small vector and largely forgotten
2. **No parallelism:** can't process tokens simultaneously — training is slow

**Attention** solves both by letting every token directly attend to every other token — the whole sequence at once.

## Intuition: what does attention do?

When encoding the word "bank" in *"I went to the river bank"*, the model should attend heavily to "river" (not "money"). Attention learns which other tokens are relevant for encoding each token.

## Scaled dot-product attention

Every token is projected into three vectors:
- **Q (Query):** what am I looking for?
- **K (Key):** what do I contain?
- **V (Value):** what do I output if selected?

$$\text{Attention}(Q, K, V) = \text{softmax}\left(\frac{QK^T}{\sqrt{d_k}}\right) V$$

The $\sqrt{d_k}$ scaling prevents the dot products from growing too large (which would push softmax into vanishing-gradient territory).

```python
import torch
import torch.nn.functional as F
import math
import matplotlib.pyplot as plt
import seaborn as sns

def scaled_dot_product_attention(Q, K, V, mask=None):
    """Q, K, V: (batch, seq_len, d_k)"""
    d_k = Q.shape[-1]
    scores = torch.matmul(Q, K.transpose(-2, -1)) / math.sqrt(d_k)
    if mask is not None:
        scores = scores.masked_fill(mask == 0, float('-inf'))
    weights = F.softmax(scores, dim=-1)
    return torch.matmul(weights, V), weights

# Toy example: 4 tokens, d_k=8
torch.manual_seed(42)
seq_len, d_k = 4, 8
Q = torch.randn(1, seq_len, d_k)
K = torch.randn(1, seq_len, d_k)
V = torch.randn(1, seq_len, d_k)

output, weights = scaled_dot_product_attention(Q, K, V)
print(f'Output shape: {output.shape}')   # (1, 4, 8)
print(f'Weights shape: {weights.shape}') # (1, 4, 4)
print(f'Weights (row sums to 1):\n{weights[0].detach().round(decimals=2)}')
```

```python
# Visualise the attention weights
tokens = ['The', 'river', 'bank', 'flooded']

# Simulate attention weights (what each token attends to)
attn = weights[0].detach().numpy()

fig, ax = plt.subplots(figsize=(5, 4))
sns.heatmap(attn, xticklabels=tokens, yticklabels=tokens,
            annot=True, fmt='.2f', cmap='Blues', ax=ax)
ax.set_title('Attention weights\n(row = query token, col = key token)')
ax.set_ylabel('Query')
ax.set_xlabel('Key')
plt.tight_layout()
plt.show()
```

## Causal (masked) attention

In decoder-only models (GPT), each token can only attend to *previous* tokens (no peeking at the future during training). This is enforced with a causal mask.

```python
def causal_mask(seq_len):
    # Lower triangular matrix — token i can attend to tokens 0..i
    return torch.tril(torch.ones(seq_len, seq_len)).unsqueeze(0)

mask = causal_mask(seq_len)
output_causal, weights_causal = scaled_dot_product_attention(Q, K, V, mask=mask)

print('Causal attention weights (upper triangle = 0):')
print(weights_causal[0].detach().round(decimals=2))
```

## Key takeaways

- Attention computes a **weighted average of Values** where weights come from **Query-Key similarity**
- Scaling by $\sqrt{d_k}$ prevents softmax saturation
- **Bidirectional attention** (BERT): every token attends to every other — good for understanding
- **Causal attention** (GPT): each token only attends to past — required for autoregressive generation
