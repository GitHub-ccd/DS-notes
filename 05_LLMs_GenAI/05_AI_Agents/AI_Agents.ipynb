---

> **2026 state of agents:** Function calling is now stable across all major model APIs and is the foundation of every production agent. LangGraph and the Anthropic/OpenAI agent SDKs have matured into the primary build surfaces. AutoGen/CrewAI are popular for research and prototyping but less common in production. The biggest unsolved problem is *reliability* — agents still fail on long multi-step tasks at rates that require human-in-the-loop checkpoints.
