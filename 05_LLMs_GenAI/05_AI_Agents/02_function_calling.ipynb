# 02 — Function Calling

Function calling (also called **tool use**) allows LLMs to request execution of external functions by outputting structured JSON describing the function name and arguments. The caller executes the function and returns the result.

## Anthropic tool use API

```python
import anthropic
import json

client = anthropic.Anthropic()

# Define a tool with JSON Schema for the input
search_tool = {
    'name': 'search_products',
    'description': 'Search for products in the catalog. Returns a list of matching products with prices.',
    'input_schema': {
        'type': 'object',
        'properties': {
            'query': {
                'type': 'string',
                'description': 'Search query',
            },
            'max_price': {
                'type': 'number',
                'description': 'Maximum price filter (optional)',
            },
            'category': {
                'type': 'string',
                'enum': ['electronics', 'clothing', 'books', 'all'],
                'description': 'Product category',
            },
        },
        'required': ['query'],
    },
}

# Simulate tool execution
def search_products(query, max_price=None, category='all'):
    catalog = [
        {'name': 'Laptop Pro 15"', 'price': 1299, 'category': 'electronics'},
        {'name': 'Wireless Mouse', 'price': 45, 'category': 'electronics'},
        {'name': 'Python Crash Course', 'price': 35, 'category': 'books'},
        {'name': 'USB-C Hub', 'price': 55, 'category': 'electronics'},
    ]
    results = [p for p in catalog if query.lower() in p['name'].lower()]
    if max_price:
        results = [p for p in results if p['price'] <= max_price]
    if category != 'all':
        results = [p for p in results if p['category'] == category]
    return results

# --- Step 1: Ask the LLM ---
response = client.messages.create(
    model='claude-haiku-4-5-20251001',
    max_tokens=512,
    tools=[search_tool],
    messages=[{'role': 'user', 'content': 'Find me electronics under $100'}],
)

print('Stop reason:', response.stop_reason)
for block in response.content:
    print(f'Block type: {block.type}')
    if block.type == 'tool_use':
        print(f'  Tool: {block.name}')
        print(f'  Input: {block.input}')
```

```python
# --- Step 2: Execute the tool and return result ---
tool_use_block = next(b for b in response.content if b.type == 'tool_use')

# Execute
tool_result = search_products(**tool_use_block.input)
print('Tool result:', json.dumps(tool_result, indent=2))

# --- Step 3: Send result back to LLM ---
final_response = client.messages.create(
    model='claude-haiku-4-5-20251001',
    max_tokens=512,
    tools=[search_tool],
    messages=[
        {'role': 'user', 'content': 'Find me electronics under $100'},
        {'role': 'assistant', 'content': response.content},
        {'role': 'user', 'content': [{
            'type': 'tool_result',
            'tool_use_id': tool_use_block.id,
            'content': json.dumps(tool_result),
        }]},
    ],
)

print('\nFinal answer:')
print(final_response.content[0].text)
```

## Parallel tool calls

```python
# Claude can call multiple tools in one turn
tools = [
    {
        'name': 'get_stock_price',
        'description': 'Get current stock price for a ticker symbol.',
        'input_schema': {
            'type': 'object',
            'properties': {'ticker': {'type': 'string'}},
            'required': ['ticker'],
        },
    },
]

def get_stock_price(ticker):
    prices = {'AAPL': 185.50, 'GOOGL': 140.20, 'MSFT': 415.30, 'AMZN': 185.00}
    return {'ticker': ticker, 'price': prices.get(ticker, 'Unknown'), 'currency': 'USD'}

response = client.messages.create(
    model='claude-haiku-4-5-20251001',
    max_tokens=512,
    tools=tools,
    messages=[{'role': 'user', 'content': 'What are the current prices of AAPL, GOOGL, and MSFT?'}],
)

# Collect all parallel tool calls
tool_calls = [b for b in response.content if b.type == 'tool_use']
print(f'{len(tool_calls)} parallel tool calls:')

tool_results = []
for call in tool_calls:
    result = get_stock_price(**call.input)
    print(f'  {call.name}({call.input}) → {result}')
    tool_results.append({
        'type': 'tool_result',
        'tool_use_id': call.id,
        'content': json.dumps(result),
    })
```

```python
# Return all results at once
final = client.messages.create(
    model='claude-haiku-4-5-20251001',
    max_tokens=512,
    tools=tools,
    messages=[
        {'role': 'user', 'content': 'What are the current prices of AAPL, GOOGL, and MSFT?'},
        {'role': 'assistant', 'content': response.content},
        {'role': 'user', 'content': tool_results},
    ],
)
print(final.content[0].text)
```

## Tool design best practices

| Principle | Why |
|-----------|-----|
| Write clear descriptions | The LLM decides which tool to call based on descriptions |
| Use `enum` for fixed choices | Prevents hallucinated parameter values |
| Mark required vs optional | Reduces LLM confusion about when to use fields |
| Return structured JSON | Easier for the LLM to parse and reason about |
| Validate inputs before execution | Never trust LLM-generated tool arguments blindly |
| Handle errors and return them | The LLM can recover from errors if it gets clear feedback |
