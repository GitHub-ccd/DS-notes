# 04 — LangGraph

**LangGraph** is a graph-based agent framework from LangChain. Instead of a linear chain, you define a **state graph** — nodes are actions (LLM calls, tool calls, logic), edges are transitions. This enables:

- **Conditional branching** — route based on LLM output
- **Cycles** — loop until a condition is met
- **Human-in-the-loop** — pause for approval before acting
- **Parallel execution** — fan out to multiple agents

## Basic LangGraph agent

```python
# pip install langgraph langchain-anthropic
from typing import Annotated, TypedDict
from langgraph.graph import StateGraph, END
from langgraph.prebuilt import ToolNode
from langchain_anthropic import ChatAnthropic
from langchain.tools import tool
import math

@tool
def calculate(expression: str) -> str:
    """Evaluate a mathematical expression. Use Python math syntax."""
    safe_globals = {'__builtins__': {}, 'sqrt': math.sqrt, 'pi': math.pi}
    try:
        return str(eval(expression, safe_globals))
    except Exception as e:
        return f'Error: {e}'

@tool
def lookup_unit_conversion(value: float, from_unit: str, to_unit: str) -> str:
    """Convert between common units. Supports: km/miles, kg/lbs, celsius/fahrenheit."""
    conversions = {
        ('km', 'miles'): lambda x: x * 0.621371,
        ('miles', 'km'): lambda x: x * 1.60934,
        ('kg', 'lbs'): lambda x: x * 2.20462,
        ('lbs', 'kg'): lambda x: x / 2.20462,
        ('celsius', 'fahrenheit'): lambda x: x * 9/5 + 32,
        ('fahrenheit', 'celsius'): lambda x: (x - 32) * 5/9,
    }
    fn = conversions.get((from_unit.lower(), to_unit.lower()))
    if fn:
        return f'{value} {from_unit} = {fn(value):.4f} {to_unit}'
    return f'No conversion found for {from_unit} to {to_unit}'

tools = [calculate, lookup_unit_conversion]
llm = ChatAnthropic(model='claude-haiku-4-5-20251001', temperature=0).bind_tools(tools)

print('LangGraph tools:', [t.name for t in tools])
```

```python
from langgraph.graph.message import add_messages

# State is a TypedDict — each node reads/writes to this shared state
class AgentState(TypedDict):
    messages: Annotated[list, add_messages]  # add_messages = append, not replace

# Node 1: Call the LLM
def call_llm(state: AgentState):
    response = llm.invoke(state['messages'])
    return {'messages': [response]}

# Routing function: go to tools or end?
def should_continue(state: AgentState):
    last_message = state['messages'][-1]
    if last_message.tool_calls:
        return 'tools'   # LLM requested a tool call
    return END           # LLM is done

# Build the graph
graph = StateGraph(AgentState)

graph.add_node('llm', call_llm)
graph.add_node('tools', ToolNode(tools))  # prebuilt: executes tool calls from messages

graph.set_entry_point('llm')
graph.add_conditional_edges('llm', should_continue)   # llm → tools or END
graph.add_edge('tools', 'llm')                        # tools → llm (loop back)

app = graph.compile()
print('Graph compiled successfully')
```

```python
from langchain_core.messages import HumanMessage

# Run the agent
result = app.invoke({
    'messages': [HumanMessage(content='Convert 100 km to miles, then calculate the square root of that result.')]
})

# Print the conversation
for msg in result['messages']:
    role = msg.__class__.__name__
    if hasattr(msg, 'content') and msg.content:
        print(f'[{role}]: {str(msg.content)[:200]}')
```

## Checkpointing — save and resume state

```python
from langgraph.checkpoint.memory import MemorySaver

# Checkpointer saves state after each node — enables pause/resume and multi-turn sessions
memory = MemorySaver()
app_with_memory = graph.compile(checkpointer=memory)

config = {'configurable': {'thread_id': 'session_1'}}

# First turn
r1 = app_with_memory.invoke(
    {'messages': [HumanMessage(content='What is 7 * 8?')]},
    config=config,
)
print('Turn 1:', r1['messages'][-1].content)

# Second turn — same thread_id continues the conversation
r2 = app_with_memory.invoke(
    {'messages': [HumanMessage(content='Now multiply that result by 2.')]},
    config=config,
)
print('Turn 2:', r2['messages'][-1].content)
```

## Human-in-the-loop — interrupt before tool execution

```python
from langgraph.checkpoint.memory import MemorySaver

# interrupt_before=['tools'] pauses execution before running any tool
app_hitl = graph.compile(
    checkpointer=MemorySaver(),
    interrupt_before=['tools'],   # pause before tool execution
)

config2 = {'configurable': {'thread_id': 'hitl_session'}}

# Run until interrupt
state = app_hitl.invoke(
    {'messages': [HumanMessage(content='Convert 50 kg to lbs')]},
    config=config2,
)

# Check what the agent wants to do (tool call is in the last message)
pending = state['messages'][-1]
if hasattr(pending, 'tool_calls') and pending.tool_calls:
    print('Agent wants to call:', pending.tool_calls[0]['name'])
    print('With args:', pending.tool_calls[0]['args'])
    print('\n[Human approves — resuming...]')
    
    # Resume — pass None to continue without adding new messages
    final = app_hitl.invoke(None, config=config2)
    print('Result:', final['messages'][-1].content)
```

## LangGraph vs AgentExecutor

| Feature | LangGraph | AgentExecutor |
|---------|-----------|---------------|
| Control flow | Explicit graph | Implicit loop |
| Branching | Conditional edges | Limited |
| State management | TypedDict, explicit | Message list |
| Human-in-the-loop | First-class (`interrupt_before`) | Hacky workarounds |
| Multi-agent | Natural (multiple nodes) | Not supported |
| Checkpointing | Built-in | Not supported |

For production agents, prefer LangGraph over `AgentExecutor`.
