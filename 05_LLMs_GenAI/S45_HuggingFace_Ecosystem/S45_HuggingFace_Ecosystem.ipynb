# Section 45 — HuggingFace Ecosystem

The complete HuggingFace toolchain. (Entry point: [MOD_4/S39_14](../../../MOD_4/S39_NLP/S39_14_transformers_huggingface.ipynb) — covers `pipeline` API and basic fine-tuning. This section goes deeper.)

[S45_01_huggingface_hub.ipynb](./S45_01_huggingface_hub.ipynb)  
Navigating the Hub: finding models, reading model cards, pushing to Hub, using Spaces and Gradio for demos.

[S45_02_tokenizers.ipynb](./S45_02_tokenizers.ipynb)  
BPE, WordPiece, SentencePiece — how text becomes token IDs. Special tokens, padding, truncation, fast tokenizers.

[S45_03_datasets_library.ipynb](./S45_03_datasets_library.ipynb)  
`datasets` library: loading from Hub, mapping transforms, filtering, batched processing, memory-mapped Arrow format, streaming for large datasets.

[S45_04_peft_lora.ipynb](./S45_04_peft_lora.ipynb)  
PEFT library: LoRA configuration, adding adapters to a base model, training only adapter weights, merging adapters back for deployment.

[S45_05_accelerate.ipynb](./S45_05_accelerate.ipynb)  
`accelerate` for device-agnostic training: single GPU → multi-GPU → mixed precision (fp16/bf16) with zero code changes.

---

> **Key libraries:** `transformers`, `datasets`, `peft`, `accelerate`, `evaluate`, `trl` — all from HuggingFace. Install with `pip install transformers datasets peft accelerate evaluate trl`.
