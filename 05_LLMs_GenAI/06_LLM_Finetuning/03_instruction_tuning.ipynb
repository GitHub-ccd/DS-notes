# 03 — Instruction Tuning

**Instruction tuning** (also called Supervised Fine-Tuning / SFT) trains a base LLM on (instruction, response) pairs so it follows directions instead of just completing text. This is the step that turns a raw pre-trained model into a useful assistant.

## Base model vs instruction-tuned model

```python
# Illustrate the difference in behavior with prompts
import anthropic

client = anthropic.Anthropic()

# A base model would just continue the text:
# "Translate 'hello' to French: bonjour. Translate 'goodbye' to French:"
# (text completion, not instruction following)

# An instruction-tuned model responds to commands:
result = client.messages.create(
    model='claude-haiku-4-5-20251001',
    max_tokens=128,
    messages=[{'role': 'user', 'content': "Translate 'hello' to French."}],
)
print('Instruction-tuned response:', result.content[0].text)

print('''
Base model behavior (GPT-2 style):
  Input:  "Translate 'hello' to French."
  Output: "Translate 'goodbye' to Spanish. Translate 'thank you' to German."
  (continues the pattern instead of translating)

Instruction-tuned behavior:
  Input:  "Translate 'hello' to French."
  Output: "Bonjour"
  (follows the instruction)
''')
```

## Creating instruction tuning data

```python
# Three sources of instruction data:
# 1. Human-written (highest quality, expensive)
# 2. Templated from existing NLP datasets
# 3. LLM-generated (Self-Instruct, Alpaca, Evol-Instruct)

# Self-Instruct: use a strong LLM to generate (instruction, response) pairs
def generate_instruction_pair(topic, client):
    msg = client.messages.create(
        model='claude-haiku-4-5-20251001',
        max_tokens=512,
        system='Generate a training example for fine-tuning a data science assistant. Respond with JSON only: {"instruction": str, "response": str}',
        messages=[{'role': 'user', 'content': f'Generate an instruction-response pair about: {topic}'}],
    )
    import json
    return json.loads(msg.content[0].text)

topics = ['cross-validation', 'regularization', 'random forests']
examples = []
for topic in topics:
    pair = generate_instruction_pair(topic, client)
    examples.append(pair)
    print(f'Topic: {topic}')
    print(f'  Q: {pair["instruction"][:80]}')
    print(f'  A: {pair["response"][:80]}...')
    print()
```

## Instruction data quality criteria

```python
# Quality matters more than quantity — 1k high-quality > 100k low-quality

quality_rubric = {
    'Diversity': [
        'Cover different task types: QA, summarization, code, classification',
        'Vary instruction phrasing ("Write...", "Explain...", "List...", "Compare...")',
        'Include edge cases and difficult questions',
    ],
    'Response quality': [
        'Responses should be correct (obvious, but verify!)',
        'Appropriate length — not too verbose, not too terse',
        'Consistent format if format matters to you',
    ],
    'Avoid': [
        'Duplicate instructions (reduces diversity, overfits)',
        'Incorrect responses (directly harms quality)',
        'Instructions the model should refuse (include a small set for safety training)',
    ],
}

for category, items in quality_rubric.items():
    print(f'{category}:')
    for item in items:
        print(f'  • {item}')
    print()
```

## Chat templates — formatting matters

```python
# Each model family has its own chat template
# Wrong template = poor fine-tune performance

templates = {
    'Llama-3': '<|begin_of_text|><|start_header_id|>user<|end_header_id|>\n{instruction}<|eot_id|><|start_header_id|>assistant<|end_header_id|>\n{response}<|eot_id|>',
    'Mistral': '[INST] {instruction} [/INST]{response}</s>',
    'ChatML (Qwen/Phi)': '<|im_start|>user\n{instruction}<|im_end|>\n<|im_start|>assistant\n{response}<|im_end|>',
    'Alpaca': '### Instruction:\n{instruction}\n\n### Response:\n{response}',
}

instruction = 'What is dropout?'
response = 'Dropout randomly zeros neuron activations during training to prevent co-adaptation and overfitting.'

for name, template in templates.items():
    print(f'--- {name} ---')
    print(template.format(instruction=instruction, response=response))
    print()

# In practice: use tokenizer.apply_chat_template() — reads the template from the model card
print("Best practice: tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)")
```

## Training only on completions (not instructions)

```python
# Key: compute loss only on the RESPONSE tokens, not the instruction tokens
# Instructions should have loss_mask = 0

# TRL's DataCollatorForCompletionOnlyLM does this automatically
# It sets labels=-100 for instruction tokens (ignored by cross-entropy)

from transformers import AutoTokenizer
from trl import DataCollatorForCompletionOnlyLM

tokenizer = AutoTokenizer.from_pretrained('gpt2')  # for demo
tokenizer.pad_token = tokenizer.eos_token

# For Alpaca-style prompts, specify the response start token
response_template = '### Response:\n'
collator = DataCollatorForCompletionOnlyLM(response_template, tokenizer=tokenizer)

sample_text = '### Instruction:\nWhat is gradient descent?\n\n### Response:\nGradient descent minimizes loss by...'
encoded = tokenizer(sample_text, return_tensors='pt')

batch = collator([encoded['input_ids'][0].tolist()])
labels = batch['labels'][0]

instruction_mask = (labels == -100).sum().item()
response_tokens = (labels != -100).sum().item()
print(f'Instruction tokens (masked, no loss): {instruction_mask}')
print(f'Response tokens (train on these):     {response_tokens}')
```
