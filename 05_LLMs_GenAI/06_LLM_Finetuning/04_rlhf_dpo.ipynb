# 04 — RLHF and DPO

Instruction tuning (SFT) teaches a model to follow instructions. **Alignment training** goes further — it teaches the model to prefer outputs that humans rate as helpful, harmless, and honest.

## RLHF: the original alignment approach

**Reinforcement Learning from Human Feedback** (RLHF) involves three stages:

1. **Collect preference data** — for the same prompt, humans choose between two model outputs
2. **Train a reward model (RM)** — binary classifier predicting which output humans prefer
3. **RL fine-tune** — optimize the LLM with PPO to maximize the RM's reward

RLHF is expensive and unstable. It requires maintaining 3 models (SFT, RM, PPO policy) and careful reward hacking prevention.

````python
# Preference data format (used for both RLHF and DPO)
import json

preference_dataset = [
    {
        'prompt': 'Explain what a p-value is.',
        'chosen': 'A p-value is the probability of observing results at least as extreme as your data, assuming the null hypothesis is true. A p-value < 0.05 means results are unlikely (< 5% chance) under the null hypothesis.',
        'rejected': 'A p-value tells you if something is statistically significant. If it\'s less than 0.05 you reject the null hypothesis.',
    },
    {
        'prompt': 'What is the difference between precision and recall?',
        'chosen': 'Precision = TP/(TP+FP): of all positive predictions, what fraction were correct? Recall = TP/(TP+FN): of all actual positives, what fraction did we find? High precision = few false alarms. High recall = few misses.',
        'rejected': 'Precision and recall are both metrics for classification. You want both to be high.',
    },
]

for i, example in enumerate(preference_dataset):
    print(f'Example {i+1}:')
    print(f'  Prompt:   {example["prompt"]}')
    print(f'  Chosen:   {example["chosen"][:100]}...')
    print(f'  Rejected: {example["rejected"][:100]}...')
    print()
````

## DPO: Direct Preference Optimization

**DPO** (Rafailov et al. 2023) achieves the same alignment goal as RLHF without needing a separate reward model or RL training. It directly optimizes the LLM using a contrastive loss over preference pairs.

The DPO loss:
```
L_DPO = -E[log σ(β · (log π(y_w|x)/π_ref(y_w|x) - log π(y_l|x)/π_ref(y_l|x)))]
```

Where `y_w` = chosen (winner), `y_l` = rejected (loser), `β` controls the strength of deviation from the reference model.

````python
# DPO training with TRL's DPOTrainer
# pip install trl

dpo_script = '''
from transformers import AutoModelForCausalLM, AutoTokenizer, TrainingArguments
from peft import LoraConfig, get_peft_model
from trl import DPOTrainer, DPOConfig
from datasets import Dataset

model_name = "meta-llama/Llama-3.2-1B-Instruct"  # start from SFT model

model = AutoModelForCausalLM.from_pretrained(model_name, device_map="auto")
ref_model = AutoModelForCausalLM.from_pretrained(model_name, device_map="auto")  # frozen reference
tokenizer = AutoTokenizer.from_pretrained(model_name)

# Apply LoRA to policy model (not reference)
lora_config = LoraConfig(r=16, lora_alpha=32, target_modules=["q_proj","v_proj"])
model = get_peft_model(model, lora_config)

# Preference dataset
dataset = Dataset.from_list([
    {"prompt": "...", "chosen": "...", "rejected": "..."},
])

dpo_config = DPOConfig(
    output_dir="./dpo-output",
    num_train_epochs=1,
    per_device_train_batch_size=2,
    beta=0.1,              # KL penalty; lower = more divergence from reference allowed
    learning_rate=5e-5,
)

trainer = DPOTrainer(
    model=model,
    ref_model=ref_model,
    args=dpo_config,
    train_dataset=dataset,
    tokenizer=tokenizer,
)
trainer.train()
'''

print('DPO training script (requires GPU):')
print(dpo_script)
````

## Generating preference data with LLM-as-judge

````python
import anthropic
import json

client = anthropic.Anthropic()

def generate_preference_pair(prompt):
    """Generate a chosen/rejected pair using LLM-as-judge."""
    # Generate two responses at different temperatures
    responses = []
    for temp in [0.0, 1.0]:
        msg = client.messages.create(
            model='claude-haiku-4-5-20251001',
            max_tokens=256,
            temperature=temp,
            messages=[{'role': 'user', 'content': prompt}],
        )
        responses.append(msg.content[0].text)
    
    # Judge which is better
    judge_msg = client.messages.create(
        model='claude-haiku-4-5-20251001',
        max_tokens=64,
        system='You judge which response is better (more accurate, clear, complete). Reply ONLY with JSON: {"better": "A" or "B", "reason": str}',
        messages=[{'role': 'user', 'content': f'Question: {prompt}\n\nA: {responses[0]}\n\nB: {responses[1]}'}],
    )
    verdict = json.loads(judge_msg.content[0].text)
    
    chosen_idx = 0 if verdict['better'] == 'A' else 1
    rejected_idx = 1 - chosen_idx
    return {
        'prompt': prompt,
        'chosen': responses[chosen_idx],
        'rejected': responses[rejected_idx],
        'judge_reason': verdict['reason'],
    }

pair = generate_preference_pair('What is the bias-variance tradeoff?')
print('Prompt:', pair['prompt'])
print('Chosen:', pair['chosen'][:150])
print('Rejected:', pair['rejected'][:150])
print('Reason:', pair['judge_reason'])
````

## RLHF vs DPO comparison

| Aspect | RLHF (PPO) | DPO |
|--------|-----------|-----|
| Reward model | Needed (separate training) | Not needed |
| Training stability | Difficult (reward hacking) | More stable |
| Models in memory | 3 (SFT, RM, policy) | 2 (policy + reference) |
| Implementation complexity | High | Low (just a loss function) |
| Quality | State-of-the-art | Close to RLHF |
| When to use | When online feedback is available | Offline preference datasets |

> **2026 context:** DPO and its variants (IPO, SimPO, ORPO) have largely replaced PPO-based RLHF for most fine-tuning use cases. RLHF is still used by large labs for production models.
