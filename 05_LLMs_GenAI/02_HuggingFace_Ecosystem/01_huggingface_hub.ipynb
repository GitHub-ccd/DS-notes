The Hub (huggingface.co) is the central registry for open-source ML models, datasets, and demo apps (Spaces). As of 2026 it hosts 900k+ models and 200k+ datasets.

## Finding and loading models

```python
# pip install huggingface_hub transformers
from huggingface_hub import HfApi, list_models

api = HfApi()

# Search for sentence-transformers models
models = list(list_models(filter='sentence-transformers', limit=5, sort='downloads', direction=-1))
for m in models:
    print(f'{m.modelId:50} downloads: {m.downloads:,}')
```

## Reading model cards

Before using any model, read its model card at `huggingface.co/<org>/<model>`. Key things to check:
- **Intended uses / limitations** — what was it trained for?
- **Training data** — what corpus? Any licensing issues?
- **Evaluation results** — benchmark scores on relevant tasks
- **License** — MIT/Apache 2.0 (permissive) vs Llama Community License (restricted commercial use)

## Spaces — Gradio demos

```python
# pip install gradio
# A Space is a Gradio app hosted on HuggingFace — deploy by pushing app.py to a Space repo

import gradio as gr
from transformers import pipeline

classifier = pipeline('sentiment-analysis')

def predict(text):
    result = classifier(text)[0]
    return f"{result['label']} (confidence: {result['score']:.2%})"

# Launch locally — deploy to Spaces by pushing this file
demo = gr.Interface(fn=predict, inputs='text', outputs='text', title='Sentiment Classifier')
# demo.launch()  # uncomment to run locally
print('Gradio app defined — call demo.launch() to run locally, or push to HF Spaces')
```

## The `evaluate` library

`pip install evaluate` — HuggingFace's metric library. Provides BLEU, ROUGE, accuracy, F1, and many more.

```python
import evaluate

accuracy = evaluate.load('accuracy')
result = accuracy.compute(predictions=[1, 0, 1, 1], references=[1, 0, 0, 1])
print(result)  # {'accuracy': 0.75}

f1 = evaluate.load('f1')
print(f1.compute(predictions=[1, 0, 1, 1], references=[1, 0, 0, 1], average='macro'))
```
