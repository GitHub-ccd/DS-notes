# Pandas Data Access — .loc, .iloc, and Boolean Indexing

## Introduction

Selecting the right rows and columns is the most frequent operation in data analysis. Pandas provides three complementary mechanisms: `loc` (label-based), `iloc` (position-based), and boolean indexing (condition-based). Knowing which to use — and when — is essential.

## Objectives

You will be able to:

* Select rows and columns using `.loc[]` with labels
* Select rows and columns using `.iloc[]` with integer positions
* Filter rows using boolean masks and compound conditions
* Use `.query()` for readable filtering
* Modify selected subsets of a DataFrame

---

## Setup

```python
import pandas as pd
import numpy as np

np.random.seed(42)
df = pd.DataFrame({
    'name':   ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve', 'Frank'],
    'dept':   ['Eng', 'Mkt', 'Eng', 'DS', 'Mkt', 'DS'],
    'age':    [25, 32, 28, 35, 22, 40],
    'salary': [65000, 82000, 70000, 95000, 58000, 105000],
    'rating': [4.2, 3.8, 4.5, 4.9, 3.5, 4.7],
})
print(df)
```

---

## `.loc[]` — Label-based Selection

```python
# Syntax: df.loc[row_labels, col_labels]

# Single row (returns Series)
print(df.loc[2])           # row with index label 2

# Single cell
print(df.loc[2, 'name'])   # Charlie

# Multiple rows
print(df.loc[[0, 2, 4]])

# Slice of rows (INCLUSIVE of both endpoints — unlike Python slicing)
print(df.loc[1:3])

# All rows, specific columns
print(df.loc[:, ['name', 'salary']])

# Slice of rows AND columns
print(df.loc[1:3, 'dept':'salary'])
```

```python
# .loc is most powerful with a named index
df_named = df.set_index('name')

print(df_named.loc['Alice'])           # Alice's row
print(df_named.loc['Bob', 'salary'])   # Bob's salary: 82000
print(df_named.loc[['Alice', 'Diana'], 'salary'])  # two employees' salaries
```

---

## `.iloc[]` — Position-based Selection

```python
# Syntax: df.iloc[row_positions, col_positions]
# Uses integer positions — like standard Python list indexing (exclusive of end)

# Single row
print(df.iloc[0])          # first row

# Single cell
print(df.iloc[0, 2])       # row 0, column 2 (age=25)

# Slice of rows
print(df.iloc[1:4])        # rows 1, 2, 3 (exclusive of 4)

# Last row
print(df.iloc[-1])

# Last 2 rows, first 3 columns
print(df.iloc[-2:, :3])

# Every other row
print(df.iloc[::2])
```

### loc vs iloc

| | `.loc[]` | `.iloc[]` |
|-|----------|-----------|
| Based on | **Labels** (index values) | **Positions** (0-based integers) |
| Row end inclusive | **Yes** (`1:3` returns 1,2,3) | **No** (`1:3` returns 1,2) |
| Use when | Index has meaningful labels | You just need row number N |

## Boolean Indexing — Filter by Condition

```python
# A boolean condition returns a Series of True/False
mask = df['salary'] > 80000
print(mask)
print(df[mask])  # keep only True rows
```

```python
# Compound conditions — use & (and) and | (or), NOT Python 'and'/'or'
# Always wrap each condition in parentheses!

# Salary > 70k AND in Engineering
eng_high = df[(df['salary'] > 70000) & (df['dept'] == 'Eng')]
print("Eng with salary > 70k:")
print(eng_high)

# Age < 30 OR rating > 4.5
young_or_excellent = df[(df['age'] < 30) | (df['rating'] > 4.5)]
print("\nYoung or excellent:")
print(young_or_excellent)

# NOT: employees who are NOT in Marketing
non_mkt = df[~(df['dept'] == 'Mkt')]
print("\nNot in Mkt:")
print(non_mkt)
```

```python
# isin() — test against a list of values
tech_depts = df[df['dept'].isin(['Eng', 'DS'])]
print(tech_depts)

# between() — inclusive range
mid_earners = df[df['salary'].between(65000, 90000)]
print(mid_earners)

# str.contains() for string matching
names_with_a = df[df['name'].str.contains('a', case=False)]
print(names_with_a)
```

---

## `.query()` — Readable String-based Filtering

```python
# .query() takes a string expression — often more readable than boolean masks

# Equivalent to df[(df['salary'] > 80000) & (df['dept'] != 'Mkt')]
result = df.query("salary > 80000 and dept != 'Mkt'")
print(result)

# Use @ to reference a Python variable
threshold = 30
young = df.query("age < @threshold")
print(young)
```

---

## Modifying Selected Subsets

```python
df_copy = df.copy()  # always work on a copy when mutating

# Give all DS employees a 10% raise
df_copy.loc[df_copy['dept'] == 'DS', 'salary'] *= 1.10

# Set a specific cell
df_copy.loc[0, 'rating'] = 4.8

print(df_copy)
```

## Summary

| Method | Syntax | When to use |
|--------|--------|-------------|
| `.loc[rows, cols]` | labels | Named index, column names |
| `.iloc[rows, cols]` | integers | Positional access, first/last N rows |
| Boolean mask | `df[df['col'] > x]` | Condition-based filtering |
| `.query()` | string expression | Complex filters, readable code |
| `.isin()` | `df['col'].isin([...])` | Membership in a list of values |
