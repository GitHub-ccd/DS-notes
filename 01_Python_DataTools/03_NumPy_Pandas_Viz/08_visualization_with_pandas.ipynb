# Visualization with Pandas

## Introduction

Pandas has a built-in `.plot()` method that wraps Matplotlib for quick, DataFrame-native visualizations. It's ideal for rapid EDA — you can go from a DataFrame to a useful chart in one line. For publication-quality or highly customized plots, you'll use Matplotlib or Seaborn directly (covered in the next two notebooks).

## Objectives

You will be able to:

* Create line, bar, histogram, scatter, and box plots from DataFrames using `.plot()`
* Customize titles, labels, and colors
* Use `.plot()` subplots for multi-panel figures
* Choose the right chart type for the data and question

---

## Setup

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline

np.random.seed(42)

# Monthly sales data
months = pd.date_range('2024-01', periods=12, freq='ME')
sales = pd.DataFrame({
    'month':   months,
    'revenue': np.random.normal(100000, 15000, 12).round(-2).astype(int),
    'units':   np.random.randint(800, 1500, 12),
    'region':  ['North', 'South', 'East', 'West'] * 3,
}).set_index('month')

# Employee data
n = 200
emp = pd.DataFrame({
    'dept':     np.random.choice(['Engineering', 'Marketing', 'DS', 'Sales'], n),
    'salary':   np.random.normal(75000, 18000, n).clip(35000, 140000).round(-2).astype(int),
    'age':      np.random.randint(22, 60, n),
    'experience': np.random.randint(1, 25, n),
})

print(sales.head())
print(emp.head())
```

---

## Line Plot — Trends Over Time

```python
# Basic line plot — index becomes x-axis
sales['revenue'].plot(figsize=(10, 4), title='Monthly Revenue 2024')
plt.ylabel('Revenue ($)')
plt.xlabel('')
plt.tight_layout()
plt.show()
```

```python
# Multiple lines — plot all numeric columns
# Normalize to make them comparable on the same axis
sales_norm = sales[['revenue', 'units']].apply(lambda col: (col - col.min()) / (col.max() - col.min()))
sales_norm.plot(figsize=(10, 4), title='Revenue and Units — Normalized (0–1)')
plt.ylabel('Normalized value')
plt.tight_layout()
plt.show()
```

---

## Bar Charts — Comparing Categories

```python
# Average salary by department — vertical bar
dept_salary = emp.groupby('dept')['salary'].mean().sort_values(ascending=False)
dept_salary.plot(kind='bar', figsize=(8, 4), title='Average Salary by Department', rot=0)
plt.ylabel('Salary ($)')
plt.tight_layout()
plt.show()
```

```python
# Horizontal bar chart — easier to read long category labels
dept_salary.plot(kind='barh', figsize=(8, 4), title='Average Salary by Department')
plt.xlabel('Salary ($)')
plt.tight_layout()
plt.show()
```

```python
# Stacked bar — headcount by dept
dept_counts = emp.groupby(['dept']).size()
dept_counts.plot(kind='bar', color='steelblue', figsize=(8, 4),
                 title='Headcount by Department', rot=0)
plt.ylabel('Employees')
plt.tight_layout()
plt.show()
```

---

## Histogram — Distributions

```python
# Single column histogram
emp['salary'].plot(kind='hist', bins=20, edgecolor='black',
                  figsize=(8, 4), title='Salary Distribution')
plt.xlabel('Salary ($)')
plt.tight_layout()
plt.show()
```

```python
# All numeric columns — subplots=True creates a panel per column
emp[['salary', 'age', 'experience']].plot(
    kind='hist', subplots=True, bins=20, layout=(1, 3),
    figsize=(14, 4), edgecolor='black', legend=False
)
plt.suptitle('Distributions of Key Variables', y=1.02)
plt.tight_layout()
plt.show()
```

---

## Scatter Plot — Relationships Between Variables

```python
emp.plot(kind='scatter', x='experience', y='salary',
         figsize=(8, 5), title='Experience vs Salary',
         alpha=0.5, c='steelblue')
plt.xlabel('Years of Experience')
plt.ylabel('Salary ($)')
plt.tight_layout()
plt.show()
```

---

## Box Plot — Distribution by Group

```python
# Box plot per department
emp.boxplot(column='salary', by='dept', figsize=(10, 5))
plt.title('Salary Distribution by Department')
plt.suptitle('')  # remove default suptitle
plt.ylabel('Salary ($)')
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```

---

## Choosing the Right Chart

| Question | Chart type |
|----------|------------|
| How does X change over time? | Line |
| How do categories compare? | Bar / Horizontal bar |
| What is the distribution of X? | Histogram |
| Is there a relationship between X and Y? | Scatter |
| How does the distribution differ across groups? | Box plot |
| What is the composition of a whole? | Pie / Stacked bar |
| How do multiple variables correlate? | Heatmap (Seaborn) |

## Summary

Pandas `.plot()` syntax: `df['col'].plot(kind='...')` or `df.plot(kind='...', x='col1', y='col2')`.

Key `kind` values: `'line'`, `'bar'`, `'barh'`, `'hist'`, `'scatter'`, `'box'`, `'pie'`, `'area'`.

Use `subplots=True` to create a panel per column. Use `figsize=(w, h)` to control size.
