# Pandas Data Structures — Series and DataFrames

## Introduction

Pandas introduces two core data structures: `Series` (a 1D labeled array) and `DataFrame` (a 2D labeled table). Both are built on top of NumPy arrays but add named indices, mixed column types, and a rich API for data manipulation.

---

## The Series

```python
import pandas as pd
import numpy as np

# From a list — auto-integer index
s1 = pd.Series([10, 20, 30, 40])
print(s1)
print(f"dtype: {s1.dtype}, shape: {s1.shape}")
```

```python
# From a dict — keys become the index
population = pd.Series({
    'California': 39_538_223,
    'Texas':      29_145_505,
    'Florida':    21_538_187,
    'New York':   20_201_249,
    'Pennsylvania': 13_002_700
})
print(population)
```

```python
# Accessing by label (index)
print(population['Texas'])           # 29145505
print(population[['California', 'Florida']])  # fancy indexing by label

# Vectorized operations — same as NumPy
print((population / 1_000_000).round(1))  # in millions

# Boolean filtering
large_states = population[population > 20_000_000]
print(large_states)
```

```python
# Index alignment — Pandas aligns on labels, not position
a = pd.Series([1, 2, 3], index=['x', 'y', 'z'])
b = pd.Series([10, 20, 30], index=['y', 'z', 'w'])
print(a + b)  # x=NaN, y=22, z=33, w=NaN
# Unmatched labels become NaN — important to understand!
```

## The DataFrame

```python
# From a dict of lists — most common creation pattern
data = {
    'name':    ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'],
    'age':     [25, 32, 28, 35, 22],
    'salary':  [65000, 82000, 70000, 95000, 58000],
    'dept':    ['Eng', 'Mkt', 'Eng', 'DS', 'Mkt'],
    'active':  [True, True, False, True, True]
}
df = pd.DataFrame(data)
print(df)
```

```python
# Core inspection methods
print(df.shape)     # (5, 5) — rows, columns
print(df.dtypes)    # data type of each column
print(df.columns.tolist())
print(df.index.tolist())
```

```python
print(df.head(3))   # first 3 rows
print(df.tail(2))   # last 2 rows
print(df.info())    # summary: dtypes, non-null counts, memory usage
print(df.describe())  # statistical summary of numeric columns
```

---

## Column Access

```python
# Single column → Series
print(df['salary'])
print(type(df['salary']))     # pandas.core.series.Series

# Multiple columns → DataFrame
print(df[['name', 'salary']])

# Attribute access (only safe when column name is a valid Python identifier with no spaces)
print(df.age)  # same as df['age']
```

---

## Creating New Columns

```python
# Vectorized — operate on entire column at once
df['monthly_salary'] = df['salary'] / 12
df['seniority'] = df['age'].apply(lambda x: 'senior' if x > 30 else 'junior')
df['salary_k'] = (df['salary'] / 1000).round(1).astype(str) + 'k'

print(df[['name', 'age', 'seniority', 'salary', 'monthly_salary', 'salary_k']])
```

---

## Setting a Custom Index

```python
# Set a meaningful column as the index
df_indexed = df.set_index('name')
print(df_indexed)
print(df_indexed.loc['Alice'])   # access by name label

# Reset back to integer index
df_reset = df_indexed.reset_index()
print(df_reset.head(2))
```

## Creating DataFrames from Other Sources

```python
# From a list of dicts (common from JSON APIs)
records = [
    {'id': 1, 'product': 'Widget', 'price': 9.99},
    {'id': 2, 'product': 'Gadget', 'price': 24.99},
    {'id': 3, 'product': 'Doohickey'},  # missing price
]
df_products = pd.DataFrame(records)
print(df_products)  # NaN for missing price
```

```python
# From a 2D NumPy array
arr = np.random.randint(0, 100, size=(4, 3))
df_arr = pd.DataFrame(arr, columns=['score_1', 'score_2', 'score_3'])
print(df_arr)
```

## Summary

| | Series | DataFrame |
|-|--------|-----------|
| Dimensions | 1D (values + index) | 2D (rows + columns) |
| Create from | list, dict, array | dict of lists, list of dicts, array |
| Access | `s[label]` | `df['col']`, `df[['c1','c2']]` |
| Key methods | `.value_counts()`, `.sort_values()` | `.describe()`, `.info()`, `.head()` |
| Alignment | By label (index) | By label (index + columns) |
