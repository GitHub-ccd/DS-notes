## The Core Libraries

| Library | Alias | Purpose |
|---------|-------|---------|
| **NumPy** | `np` | N-dimensional arrays, vectorized math, linear algebra |
| **Pandas** | `pd` | Labeled tabular data (DataFrames, Series), data wrangling |
| **Matplotlib** | `plt` | Core plotting — full control, verbose |
| **Seaborn** | `sns` | Statistical visualization — prettier defaults, higher-level API |
| **Scikit-learn** | `sklearn` | Machine learning — classification, regression, clustering, pipelines |
| **Statsmodels** | `sm` | Statistical modeling — OLS, GLM, time series, hypothesis tests |
| **SciPy** | `scipy` | Scientific computing — optimization, signal processing, statistics |

**The relationship between them:**
```
NumPy  ──────────────────┐
  ↓ arrays              ↓
Pandas (built on NumPy) → Matplotlib / Seaborn (visualization)
  ↓                              ↓
Scikit-learn (ML on NumPy/Pandas arrays)
```

## A Taste of the Stack

A quick demonstration of all four core libraries working together.

````python
# 1. NumPy — generate and compute
np.random.seed(42)
data = np.random.normal(loc=170, scale=10, size=1000)  # heights in cm
print(f"NumPy: mean={np.mean(data):.1f}, std={np.std(data):.1f}")

# 2. Pandas — wrap in a DataFrame
df = pd.DataFrame({'height_cm': data})
df['height_m'] = df['height_cm'] / 100
print(f"\nPandas shape: {df.shape}")
print(df.describe())

# 3. Matplotlib — basic plot
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].hist(df['height_cm'], bins=30, edgecolor='black')
axes[0].set_title('Matplotlib: Height Distribution')
axes[0].set_xlabel('Height (cm)')

# 4. Seaborn — statistical plot
sns.histplot(df['height_cm'], kde=True, ax=axes[1])
axes[1].set_title('Seaborn: Height with KDE')
axes[1].set_xlabel('Height (cm)')

plt.tight_layout()
plt.show()
````
