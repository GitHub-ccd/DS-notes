# Seaborn — Statistical Visualization

## Introduction

Seaborn is a statistical visualization library built on Matplotlib. It understands DataFrames natively, produces publication-quality charts with minimal code, and includes statistical features (confidence intervals, regression lines, KDE) that Matplotlib requires manual computation for. Its figure-level functions (`relplot`, `displot`, `catplot`) also make creating facet grids trivial.

## Objectives

You will be able to:

* Use Seaborn's three plot families: relational, distributional, categorical
* Add hue, style, and size dimensions to encode additional variables
* Create faceted multi-panel plots with `col=` and `row=`
* Build pair plots and heatmaps for EDA
* Apply and customize Seaborn themes

---

## Setup and Themes

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
%matplotlib inline

# Seaborn themes
sns.set_theme(style='whitegrid', palette='muted', font_scale=1.1)

# Available themes
print("Styles:", ['darkgrid', 'whitegrid', 'dark', 'white', 'ticks'])
print("Palettes:", ['deep', 'muted', 'pastel', 'bright', 'dark', 'colorblind'])

# Built-in datasets
tips = sns.load_dataset('tips')
penguins = sns.load_dataset('penguins').dropna()
print(tips.head(3))
print(penguins.head(3))
```

---

## Relational Plots — Relationships Between Variables

```python
# scatterplot — one variable per semantic dimension: x, y, hue, size, style
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

# Basic scatter
sns.scatterplot(data=tips, x='total_bill', y='tip', ax=axes[0])
axes[0].set_title('Total Bill vs Tip')

# Encode a third variable with hue and a fourth with style
sns.scatterplot(data=tips, x='total_bill', y='tip',
                hue='smoker', style='time',
                palette={'Yes': 'tomato', 'No': 'steelblue'},
                ax=axes[1])
axes[1].set_title('Bill vs Tip — Smoker and Meal Time')

plt.tight_layout()
plt.show()
```

```python
# lineplot — time series with confidence interval automatically added
np.random.seed(0)
dates = pd.date_range('2024-01', periods=48, freq='W')
ts = pd.DataFrame({
    'date': list(dates) * 3,
    'revenue': np.random.normal(100, 15, 48 * 3) + np.tile(np.linspace(0, 30, 48), 3),
    'product': ['A'] * 48 + ['B'] * 48 + ['C'] * 48,
})

plt.figure(figsize=(10, 4))
sns.lineplot(data=ts, x='date', y='revenue', hue='product')
plt.title('Weekly Revenue by Product — CI shown as shaded band')
plt.tight_layout()
plt.show()
```

---

## Distributional Plots

```python
fig, axes = plt.subplots(1, 3, figsize=(14, 4))

# histplot — histogram with optional KDE overlay
sns.histplot(data=tips, x='total_bill', kde=True, ax=axes[0])
axes[0].set_title('Histogram + KDE')

# kdeplot — smooth density curve
sns.kdeplot(data=tips, x='total_bill', hue='time', fill=True, alpha=0.4, ax=axes[1])
axes[1].set_title('KDE by Meal Time')

# ecdfplot — empirical CDF
sns.ecdfplot(data=tips, x='total_bill', hue='day', ax=axes[2])
axes[2].set_title('Empirical CDF by Day')

plt.tight_layout()
plt.show()
```

---

## Categorical Plots

```python
fig, axes = plt.subplots(2, 2, figsize=(13, 9))

# boxplot — five-number summary
sns.boxplot(data=tips, x='day', y='total_bill', hue='smoker',
            order=['Thur', 'Fri', 'Sat', 'Sun'], ax=axes[0, 0])
axes[0, 0].set_title('Box Plot: Bill by Day and Smoker Status')

# violinplot — combines box plot with KDE
sns.violinplot(data=penguins, x='species', y='body_mass_g',
               hue='sex', split=True, ax=axes[0, 1])
axes[0, 1].set_title('Violin: Penguin Body Mass')

# barplot — mean + CI
sns.barplot(data=tips, x='day', y='tip', hue='smoker',
            order=['Thur', 'Fri', 'Sat', 'Sun'], ax=axes[1, 0])
axes[1, 0].set_title('Bar Plot: Mean Tip (with 95% CI)')

# stripplot / swarmplot — show all individual data points
sns.stripplot(data=penguins, x='species', y='flipper_length_mm',
              hue='sex', dodge=True, alpha=0.5, ax=axes[1, 1])
axes[1, 1].set_title('Strip Plot: All Individual Points')

plt.tight_layout()
plt.show()
```

---

## Heatmaps and Pair Plots

```python
# Correlation heatmap
numeric_cols = penguins.select_dtypes(include='number')
corr = numeric_cols.corr()

plt.figure(figsize=(7, 5))
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm',
            center=0, square=True, linewidths=0.5)
plt.title('Penguin Feature Correlations')
plt.tight_layout()
plt.show()
```

```python
# Pair plot — all pairwise scatter plots in a grid
g = sns.pairplot(penguins, hue='species', diag_kind='kde',
                 plot_kws={'alpha': 0.5, 's': 30})
g.fig.suptitle('Penguin Features — Pair Plot', y=1.01)
plt.show()
```

---

## Facet Grids — Multi-Panel by Category

```python
# relplot, displot, catplot are figure-level functions that support faceting

# One column per island, hue by species
g = sns.relplot(
    data=penguins, x='bill_length_mm', y='bill_depth_mm',
    col='island', hue='species', style='sex',
    height=4, aspect=0.9
)
g.set_axis_labels('Bill Length (mm)', 'Bill Depth (mm)')
g.set_titles('Island: {col_name}')
plt.show()
```

```python
# lmplot — scatter with linear regression line per group
g = sns.lmplot(
    data=penguins, x='flipper_length_mm', y='body_mass_g',
    hue='species', height=5, aspect=1.2, scatter_kws={'alpha': 0.4}
)
g.set_axis_labels('Flipper Length (mm)', 'Body Mass (g)')
plt.suptitle('Body Mass vs Flipper Length by Species', y=1.01)
plt.show()
```

## Seaborn Quick Reference

| Family | Plots | Use for |
|--------|-------|---------|
| **Relational** | `scatterplot`, `lineplot` | Relationships between two continuous variables |
| **Distributional** | `histplot`, `kdeplot`, `ecdfplot`, `rugplot` | Distribution of one variable |
| **Categorical** | `boxplot`, `violinplot`, `barplot`, `stripplot`, `swarmplot`, `countplot` | Variable distributions across categories |
| **Matrix** | `heatmap`, `clustermap` | Correlation matrices, pivot tables |
| **Multi-panel** | `pairplot`, `relplot`, `displot`, `catplot` | Faceting, multiple relationships at once |
| **Regression** | `regplot`, `lmplot`, `residplot` | Linear relationships with confidence bands |

**Key semantics:** `hue` (color by category), `style` (marker/linestyle), `size` (point size), `col`/`row` (facets)
