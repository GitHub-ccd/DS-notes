# Python Style — PEP 8

## Introduction

Python code is read far more often than it is written. PEP 8 (Python Enhancement Proposal 8) is the official style guide — a set of conventions that makes Python code readable and consistent across the entire community.

Following PEP 8 is not about aesthetics. It is about **communicating intent clearly** to anyone who reads your code — including yourself six months from now.

## Objectives

You will be able to:

* Apply PEP 8 naming conventions for variables, functions, classes, and constants
* Format code with correct indentation, line length, and whitespace
* Organize imports in the standard order
* Use automated linters to check style

## Naming Conventions

```python
# Variables and functions: snake_case (lowercase words separated by underscores)
customer_count = 42
average_revenue = 1500.75

def calculate_churn_rate(customers, churned):
    return churned / customers


# Classes: PascalCase (capitalize each word, no underscores)
class CustomerSegment:
    pass

class DataPreprocessor:
    pass


# Constants: UPPER_CASE_WITH_UNDERSCORES
MAX_ITERATIONS = 1000
DEFAULT_LEARNING_RATE = 0.01
API_BASE_URL = "https://api.example.com/v1"


# Private/internal: leading underscore (convention, not enforcement)
def _internal_helper(x):
    return x * 2


# What to avoid:
# customerCount   ← camelCase (JavaScript style)
# CustomerCount   ← looks like a class
# l, O, I         ← ambiguous single letters (look like 1, 0)
```

## Indentation and Line Length

```python
# 4 spaces per indentation level — never tabs
def process_data(data):
    for item in data:
        if item > 0:
            result = item * 2
            print(result)


# Max line length: 79 characters for code, 72 for docstrings
# Break long lines using backslash or implicit continuation inside brackets:

# Backslash continuation:
total = (first_variable
         + second_variable
         + third_variable)

# Long function call:
result = some_function(
    argument_one,
    argument_two,
    argument_three,
)
```

## Whitespace

```python
# Spaces around operators:
x = 5 + 3          # good
x=5+3               # bad

# No spaces inside brackets:
my_list = [1, 2, 3]        # good
my_list = [ 1, 2, 3 ]      # bad

# Space after comma, not before:
def f(a, b, c):    # good
    pass
def f(a ,b ,c):    # bad
    pass

# No space before colon in slices:
data[1:5]           # good
data[1 : 5]         # bad

# Spaces around = for assignment, no spaces for keyword arguments:
x = 10                          # assignment — space
my_list.sort(key=len)           # keyword arg — no space
```

## Blank Lines

```python
# 2 blank lines between top-level definitions (functions, classes)
def first_function():
    pass


def second_function():
    pass


class MyClass:
    # 1 blank line between methods inside a class
    def method_one(self):
        pass

    def method_two(self):
        pass
```

## Comments and Docstrings

```python
# Comments explain WHY, not WHAT. Good code is self-documenting.

# Bad comment — just restates the code:
x = x + 1  # increment x by 1

# Good comment — explains the non-obvious reason:
x = x + 1  # compensate for zero-indexed API response


def calculate_discount(price, rate):
    """
    Calculate discount amount for a given price and rate.

    Args:
        price (float): Original price in dollars
        rate (float): Discount rate as a decimal (0.0 to 1.0)

    Returns:
        float: Discount amount (price * rate)
    """
    return price * rate
```

## PEP 8 Quick Reference

| Rule | Convention |
|------|------------|
| Variables / functions | `snake_case` |
| Classes | `PascalCase` |
| Constants | `UPPER_CASE` |
| Indentation | 4 spaces |
| Max line length | 79 chars |
| Between top-level defs | 2 blank lines |
| Between methods | 1 blank line |
| Imports order | stdlib → 3rd party → local |
| Spaces around operators | Yes (`x = a + b`) |
| Spaces inside brackets | No (`[1, 2, 3]`) |

---

## Summary

PEP 8 is a social contract: code written to these conventions can be read and understood by any Python developer without extra mental overhead. The rules are not arbitrary — they emerged from decades of community experience about what makes code maintainable. Use `flake8` or `black` in your workflow so the style enforcement is automatic rather than something you have to remember manually.
