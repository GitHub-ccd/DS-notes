## The UI in 60 Seconds

VSCode's window is split into six named areas:

```
┌─────────────────────────────────────────────────────────┐
│  Title Bar                                              │
├──┬──────────────────────────────────────────────────────┤
│  │  Side Bar          │  Editor (tabs)                  │
│A │  (Explorer,        │                                 │
│c │   Search,          │   your code lives here          │
│t │   Git,             │                                 │
│i │   Debug,           ├─────────────────────────────────┤
│v │   Extensions)      │  Panel                          │
│i │                    │  (Terminal, Output, Problems)   │
│t │                    │                                 │
│y │                    │                                 │
│  │                    │                                 │
├──┴────────────────────┴─────────────────────────────────┤
│  Status Bar  (interpreter, git branch, errors)          │
└─────────────────────────────────────────────────────────┘
```

| Area | Shortcut to toggle | What it's for |
|------|-------------------|---------------|
| **Activity Bar** | — | Icons on the far left: switch between Explorer, Search, Git, Debug, Extensions |
| **Side Bar** | `Ctrl+B` | The panel that opens when you click an Activity Bar icon |
| **Editor** | — | Where you edit files; supports split panes (`Ctrl+\`) |
| **Panel** | `Ctrl+J` | Terminal, Output, Problems, Debug Console below the editor |
| **Status Bar** | — | Bottom strip: Python interpreter, git branch, error count, line/col |
| **Command Palette** | `Ctrl+Shift+P` | Described below — the most important shortcut in all of VSCode |

## The Command Palette — Your Most Important Tool

> **`Ctrl+Shift+P`** — opens the Command Palette

Type any word and VSCode fuzzy-searches every command, setting, and action. You never need to hunt through menus. Some examples:

| Type this... | ...to do this |
|-------------|---------------|
| `select interpreter` | Choose which Python interpreter to use |
| `format document` | Auto-format the current file |
| `toggle terminal` | Open/close the integrated terminal |
| `git commit` | Commit staged changes |
| `zen mode` | Enter distraction-free full-screen |
| `open settings json` | Edit raw settings file |
| `rename symbol` | Rename a variable across all files |

The **`>`** prefix means "run a command". Remove it and you get **Quick Open** (file search by name — same as `Ctrl+P`).

| Prefix | What it searches |
|--------|------------------|
| `>` | Commands (default in Command Palette) |
| *(none)* | Files by name |
| `@` | Symbols in current file |
| `#` | Symbols across all files |
| `:` | Go to line number |

## Multi-Cursor Editing — A Game Changer

Multi-cursor lets you type in multiple places simultaneously.

```python
# Example: rename three variables in one operation
# 1. Click on 'old_name'
# 2. Press Ctrl+D twice to select the next two occurrences
# 3. Type 'new_name' — all three update at once

old_name = 'Alice'
print(old_name)
data[old_name] = 42

# Becomes:
new_name = 'Alice'
print(new_name)
data[new_name] = 42

# For column-aligned edits: hold Alt and click/drag to place cursors
# on each line at the same column position
```

**`Ctrl+Shift+L`** selects every occurrence in the file — useful for bulk renames you'd normally do with find-replace.

## Python Setup in VSCode

### 1. Install the Python Extension

Open Extensions (`Ctrl+Shift+X`) → search **Python** → install the Microsoft Python extension. This also installs **Pylance** (type checking and IntelliSense) and **Jupyter** automatically.

### 2. Select Your Python Interpreter

```
Ctrl+Shift+P → "Python: Select Interpreter"
```

VSCode will list all interpreters it finds: system Python, Anaconda base, and any conda/venv environments. Select the one for your project. Your choice is saved per-workspace in `.vscode/settings.json`.

> The currently active interpreter always shows in the **Status Bar** bottom-left — click it to change quickly.

### 3. Create / Activate a Virtual Environment

```
Ctrl+Shift+P → "Python: Create Environment"
```

Choose **Venv** or **Conda**. VSCode creates the environment and automatically selects it. From then on, every new integrated terminal in this workspace activates the environment automatically.

### 4. Run Python Files

| Method | How |
|--------|-----|
| Run whole file | Click ▶ in top-right corner of editor, or right-click → **Run Python File in Terminal** |
| Run selected lines | Select code → `Shift+Enter` (sends to interactive REPL) |
| Run in terminal | Open terminal (`Ctrl+\``) → `python filename.py` |

## The Integrated Terminal

Open with **`Ctrl+\``** (backtick). This is a full terminal inside VSCode — no need to switch windows.

| Action | How |
|--------|-----|
| New terminal | `Ctrl+Shift+\`` |
| Switch terminals | Click the dropdown in the Terminal panel |
| Split terminal | Click the split icon in the Terminal toolbar |
| Kill terminal | Click the bin icon |

The terminal **automatically activates your selected virtual environment** when it opens — so `pip install`, `python`, and `conda` commands all use the right environment without any manual activation.

```bash
# Example workflow — everything happens inside VSCode:
pip install pandas matplotlib seaborn
python train.py
jupyter notebook  # opens browser
git status
```

## Debugging Python

The debugger is where VSCode beats Jupyter notebooks — you can pause execution on any line and inspect every variable's current state.

### Step 1 — Set a Breakpoint

Click in the **gutter** (left of the line number) → a red dot appears. Or press **`F9`** while your cursor is on the line.

### Step 2 — Start Debugging

Press **`F5`** → first time, select **Python File** as the debug config. Execution pauses at your breakpoint.

### Step 3 — Inspect and Step

| Control | Shortcut | What it does |
|---------|----------|--------------|
| Continue | `F5` | Run until next breakpoint |
| Step Over | `F10` | Execute current line, stay at same level |
| Step Into | `F11` | Jump into the function being called |
| Step Out | `Shift+F11` | Finish current function, return to caller |
| Stop | `Shift+F5` | Kill the debugger |

While paused:
- **Variables panel** (left sidebar) shows all local and global variables with their current values
- **Debug Console** lets you type any Python expression and evaluate it against the current state
- **Hover** over any variable in the editor to see its current value as a tooltip

```python
# Set a breakpoint on line 7 and run with F5
# When execution pauses, type in the Debug Console:
#   df.shape       → see the DataFrame shape
#   df.head()      → inspect first rows
#   type(result)   → check a variable's type

import pandas as pd
df = pd.read_csv('data.csv')
df = df.dropna()              # <-- set breakpoint here
result = df['price'].mean()
print(result)
```

## IntelliSense — Code Completion and Type Hints

IntelliSense (powered by **Pylance**) provides:

- **Auto-complete** — start typing, see suggestions; `Tab` or `Enter` to accept
- **Parameter hints** — when you open `(`, a tooltip shows expected arguments
- **Type inference** — Pylance infers types even without explicit annotations
- **Inline documentation** — hover over any function to see its docstring
- **Error squiggles** — red underlines for syntax errors, yellow for warnings

```python
import pandas as pd
df = pd.DataFrame({'a': [1, 2, 3]})
df.  # type this, then Ctrl+Space → see every DataFrame method with docs
```

**Go to Definition** (`F12`) is especially useful: click on any Pandas or NumPy function to jump to its source code and understand exactly what it does.

## Customising VSCode — settings.json

All settings are stored in a JSON file. Open it with:

```
Ctrl+Shift+P → "Open User Settings (JSON)"
```

Useful settings to add:

```json
{
    // Editor appearance
    "editor.fontSize": 14,
    "editor.fontFamily": "'Fira Code', Consolas, monospace",
    "editor.fontLigatures": true,
    "editor.lineHeight": 1.6,
    "editor.wordWrap": "on",
    "editor.rulers": [79, 99],

    // Format on save
    "editor.formatOnSave": true,
    "[python]": {
        "editor.defaultFormatter": "ms-python.black-formatter"
    },

    // Python
    "python.analysis.typeCheckingMode": "basic",
    "python.terminal.activateEnvironment": true,

    // File explorer
    "explorer.compactFolders": false,
    "files.exclude": {
        "**/__pycache__": true,
        "**/.ipynb_checkpoints": true
    },

    // Terminal
    "terminal.integrated.fontSize": 13,
    "terminal.integrated.defaultProfile.windows": "PowerShell"
}
```

> **Workspace settings** (`.vscode/settings.json` in your project folder) override user settings for that project only — useful for project-specific interpreters or linting rules.

## Jupyter Notebooks Inside VSCode

With the **Jupyter** extension installed, you can open `.ipynb` files directly in VSCode — no browser needed.

| Action | How |
|--------|-----|
| Run cell | `Ctrl+Enter` |
| Run cell and move to next | `Shift+Enter` |
| Run all cells | Toolbar button or `Ctrl+Shift+P → Run All` |
| Add cell above/below | `A` / `B` (in command mode) |
| Toggle cell type | `M` (markdown) / `Y` (code) |
| Select kernel | Click kernel name top-right → choose your environment |
| Variable inspector | Click **Variables** in toolbar → see all in-memory objects |

> **Pro tip:** For `.py` scripts you can add `# %%` comments to create runnable "cells" just like a notebook — VSCode will show a **Run Cell** button above each one. This is called the **Interactive Window** and is great for exploratory analysis without the overhead of a `.ipynb` file.

## Git Integration

VSCode has first-class Git support without any extensions.

| Task | Where |
|------|-------|
| See changed files | Activity Bar → **Source Control** icon (or `Ctrl+Shift+G`) |
| Stage a file | Click `+` next to the file |
| Commit | Type message → `Ctrl+Enter` |
| Push / Pull | `...` menu in Source Control panel |
| View inline diff | Click any changed file — diff appears in the editor |
| Blame annotations | **GitLens** extension → hover over any line |

The **Status Bar** shows the current branch at the bottom-left. Click it to switch branches.

VSCode's power comes from three things working together:
