## Why Version Control?

Without version control, a typical project folder looks like:
```
analysis_final.ipynb
analysis_final_v2.ipynb
analysis_FINAL_USE_THIS.ipynb
analysis_FINAL_USE_THIS_fixed.ipynb
```

With Git, you have one file and a full, navigable history of every change.

Git also enables:
- **Collaboration** — multiple people editing the same codebase without overwriting each other's work
- **Experimentation** — try a new approach in a branch; discard it if it doesn't work
- **Blame and audit** — find exactly when and why something changed
- **Recovery** — restore any previous version of any file

## Git Concepts

| Concept | Description |
|---------|-------------|
| **Repository (repo)** | A directory tracked by Git, containing all project files + history |
| **Commit** | A snapshot of the project at a point in time |
| **Working directory** | Your actual files on disk |
| **Staging area (index)** | Files you've marked to include in the next commit |
| **HEAD** | Pointer to the current commit (usually the latest) |
| **Remote** | A copy of the repo on a server (GitHub, GitLab) |

**The three states of a file in Git:**
```
Modified  →  Staged  →  Committed
(working     (index)    (repo history)
 directory)
```

## Initial Setup

```bash
# Tell Git who you are (done once, globally)
git config --global user.name "Your Name"
git config --global user.email "you@example.com"

# Set default branch name to 'main'
git config --global init.defaultBranch main

# Verify
git config --list
```

## Core Workflow

### Starting a repository
```bash
# Initialize a new repo in the current directory
git init

# OR clone an existing repo from GitHub
git clone https://github.com/username/repo-name.git
cd repo-name
```

### Making changes and committing
```bash
# Check what's changed
git status

# See exactly what changed in files
git diff
git diff --staged   # changes already staged

# Stage files for commit
git add filename.py           # stage a specific file
git add .                     # stage all changes (use with care)
git add -p                    # interactively stage chunks

# Commit with a message
git commit -m "Add exploratory analysis notebook"

# Unstage a file (undo git add, without losing changes)
git restore --staged filename.py

# Discard changes in working directory (DESTRUCTIVE — cannot be undone)
git restore filename.py
```

### Viewing history
```bash
# View commit history
git log
git log --oneline           # compact view
git log --oneline --graph   # with branch visualization

# See changes in a specific commit
git show abc1234            # where abc1234 is the commit hash
```

## Working with Remotes (GitHub)

```bash
# Add a remote named 'origin'
git remote add origin https://github.com/username/repo.git

# Push commits to GitHub
git push -u origin main    # -u sets upstream, only needed first time
git push                   # subsequent pushes

# Pull latest changes from remote
git pull

# Fetch without merging (safer, lets you inspect first)
git fetch
git diff origin/main

# View configured remotes
git remote -v
```

## Writing Good Commit Messages

A good commit message is a one-line summary (≤50 chars) of *what changed and why*:

```
# Good:
Add cross-validation to model evaluation
Fix off-by-one error in IQR calculation
Remove API key that was accidentally committed

# Bad:
update stuff
fix
wip
asdfgh
```

Commit messages are for your future self and teammates. Three months from now, "add feature X because Y requirement changed" is invaluable; "update" is useless.

## Quick Reference — Git Cheat Sheet

| Command | What it does |
|---------|-------------|
| `git init` | Create new local repo |
| `git clone <url>` | Clone remote repo |
| `git status` | Show modified/staged files |
| `git diff` | Show unstaged changes |
| `git add <file>` | Stage file |
| `git commit -m "msg"` | Commit staged changes |
| `git log --oneline` | View compact history |
| `git push` | Push to remote |
| `git pull` | Pull from remote |
| `git restore --staged <f>` | Unstage file |
| `git restore <file>` | Discard working directory changes |
