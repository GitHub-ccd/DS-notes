# Git Branching and Collaboration

## Introduction

Branches are Git's killer feature for team collaboration. They let you develop features, fix bugs, or run experiments in complete isolation — without affecting the stable main codebase — then merge your work back when it's ready.

## Objectives

You will be able to:

* Create, switch, and delete branches
* Merge branches and resolve conflicts
* Use `git stash` to temporarily shelve work
* Follow a GitHub Pull Request workflow
* Apply the two golden rules of Git collaboration

---

## The Two Golden Rules

1. **Keep `main` (or `master`) functional at all times.** `main` is what other people clone. Broken `main` blocks the whole team.
2. **Keep lines of work separate.** Each feature, bug fix, or experiment gets its own branch.

---

## Branches

A branch is a lightweight pointer to a specific commit. When you create a branch, you get a copy of the current state to work on independently.

```
main:     A --- B --- C
                       \
feature:               D --- E
```

### Working with branches

```bash
# List all branches (* = current)
git branch

# Create a new branch
git branch feature/add-visualization

# Switch to a branch
git checkout feature/add-visualization

# Create and switch in one step (preferred)
git checkout -b feature/add-visualization
# or using newer syntax:
git switch -c feature/add-visualization

# Delete a branch (after merging)
git branch -d feature/add-visualization

# Force delete (unmerged branch)
git branch -D feature/add-visualization
```

### Naming conventions

| Pattern | Use |
|---------|-----|
| `feature/description` | New feature |
| `fix/bug-description` | Bug fix |
| `chore/cleanup-task` | Maintenance |
| `experiment/try-xgboost` | Exploration |

## Merging

When your work is done on a branch, merge it back:

```bash
# Switch back to main first
git checkout main

# Pull latest main from remote
git pull

# Merge feature branch into main
git merge feature/add-visualization

# Push the updated main
git push
```

---

## Merge Conflicts

A conflict occurs when the same lines were changed differently on two branches. Git stops and asks you to resolve it manually.

```
<<<<<<< HEAD (current branch — usually main)
result = model.predict(X_test)
=======
result = model.predict_proba(X_test)
>>>>>>> feature/add-probabilities
```

**To resolve:**
1. Open the conflicted file
2. Delete the `<<<<<<<`, `=======`, `>>>>>>>` markers
3. Keep the correct code (or combine both)
4. Stage and commit:

```bash
git add conflicted_file.py
git commit -m "Merge feature/add-probabilities — use predict_proba"
```

**Pro tip:** Run `git status` during a conflict — it lists exactly which files need resolution.

## Git Stash — Shelving Work in Progress

You're mid-feature and urgently need to switch branches to fix a bug, but your work isn't ready to commit. Use `stash`:

```bash
# Stash current changes (saves them, reverts to last commit)
git stash

# ... do your urgent work on another branch ...

# Come back and restore stashed changes
git stash pop

# List all stashes
git stash list

# Apply a specific stash (without removing it)
git stash apply stash@{1}

# Discard a stash
git stash drop
```

## The GitHub Pull Request Workflow

In teams, you don't merge directly to `main` — you open a **Pull Request (PR)** for review:

```
1. Fork or clone the repo
2. Create a feature branch
         git checkout -b feature/my-change
3. Make commits on your branch
4. Push branch to GitHub
         git push -u origin feature/my-change
5. Open a Pull Request on GitHub (browser)
6. Teammates review and comment
7. Address feedback with new commits
8. PR is approved and merged
9. Delete the feature branch
```

Even working solo, this pattern keeps your commit history clean and reviewable.

## Viewing Remote Branches

```bash
# See all remote branches
git branch -r

# See all branches (local + remote)
git branch -a

# Pull a remote branch locally
git checkout -b feature/remote-branch origin/feature/remote-branch
```

## Common Scenarios

### Undo the last commit (keep changes staged)
```bash
git reset --soft HEAD~1
```

### Undo the last commit (keep changes in working directory)
```bash
git reset HEAD~1
```

### See who changed a line (blame)
```bash
git blame filename.py
```

### Find when a bug was introduced
```bash
git bisect start
git bisect bad             # current commit is broken
git bisect good abc1234    # this commit was fine
# Git checks out commits to test until it finds the culprit
git bisect reset
```
