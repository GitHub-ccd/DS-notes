# Bash and the Command Line

## Introduction

The command line (also called the shell, terminal, or Bash) is a text-based interface for interacting with your computer. Before GUIs existed, everything was done this way. For data scientists, the command line is indispensable — it's where you run scripts, manage files, use Git, and SSH into remote servers.

## Objectives

You will be able to:

* Navigate the filesystem using the command line
* Create, move, copy, and delete files and folders
* Understand file paths (absolute vs relative)
* Use tab completion and command history
* Chain commands with pipes

---

## Opening a Terminal

- **Mac:** Applications → Utilities → Terminal, or press `Cmd+Space` and type "Terminal"
- **Windows:** Open Git Bash (installed with Git) or use WSL (Windows Subsystem for Linux)
- **Linux:** `Ctrl+Alt+T`

You'll see a **prompt** that looks like:
```
username@computername:~/current/directory$
```
The `$` indicates you can type a command.

## Essential Navigation Commands

| Command | Name | What it does | Example |
|---------|------|-------------|------|
| `pwd` | Print Working Directory | Show current location | `pwd` → `/home/user/DS-notes` |
| `ls` | List | Show files in current directory | `ls -la` (detailed, show hidden) |
| `cd` | Change Directory | Move to a directory | `cd Documents/DS-notes` |
| `cd ..` | | Go up one level | `cd ..` |
| `cd ~` | | Go to home directory | `cd ~` |
| `cd -` | | Go to previous directory | `cd -` |

```bash
# Practical example — navigate to this repo
pwd                          # see where you are
ls                           # list what's here
cd DS-notes                  # enter the repo
ls -la                       # list including hidden files (.git, .env)
```

## File and Directory Management

| Command | What it does | Example |
|---------|-------------|------|
| `mkdir dirname` | Create directory | `mkdir new_project` |
| `touch filename` | Create empty file | `touch notes.md` |
| `cp src dest` | Copy file | `cp data.csv data_backup.csv` |
| `cp -r src dest` | Copy directory | `cp -r assets/ assets_backup/` |
| `mv src dest` | Move or rename | `mv old_name.py new_name.py` |
| `rm filename` | Delete file | `rm temp.txt` |
| `rm -r dirname` | Delete directory | `rm -r old_project/` |
| `cat file` | Print file contents | `cat README.md` |
| `head -n 5 file` | First 5 lines | `head -n 5 data.csv` |
| `tail -n 5 file` | Last 5 lines | `tail -n 5 log.txt` |

> **Danger:** `rm -r` deletes permanently with no Recycle Bin. Double-check before running.

## File Paths

**Absolute path** — starts from root, works from anywhere:
```bash
/home/user/DS-notes/01_Python_DataTools/README.md
```

**Relative path** — relative to current directory:
```bash
./01_Python_DataTools/README.md   # ./ = current directory
../README.md                      # ../ = parent directory
```

## Productivity Tips

### Tab Completion
Press `Tab` to autocomplete file and directory names. Press `Tab` twice to see all options.
```bash
cd DS-n<Tab>   # autocompletes to DS-notes/
```

### Command History
Press `↑` / `↓` to navigate previous commands. Use `Ctrl+R` to search history.

### Pipes
The `|` operator passes output of one command as input to another:
```bash
ls -la | grep ".ipynb"   # list only .ipynb files
cat data.csv | head -20  # first 20 lines of a CSV
```

### Other useful shortcuts
| Shortcut | Effect |
|----------|--------|
| `Ctrl+C` | Cancel running command |
| `Ctrl+L` | Clear terminal |
| `Ctrl+A` | Move cursor to start of line |
| `Ctrl+E` | Move cursor to end of line |
| `!!` | Repeat last command |

## Common Data Science Shell Patterns

```bash
# Count lines in a CSV (including header)
wc -l data/train.csv

# Find all Jupyter notebooks in subdirectories
find . -name "*.ipynb" -not -path "*/.ipynb_checkpoints/*"

# Search for a string across all Python files
grep -r "import pandas" .

# Check disk usage of a folder
du -sh data/

# Download a file
curl -O https://example.com/dataset.csv
wget https://example.com/dataset.csv
```
