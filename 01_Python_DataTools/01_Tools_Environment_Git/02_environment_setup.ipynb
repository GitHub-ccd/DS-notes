## Step 3 — Virtual Environments

### Why virtual environments?

Different projects require different package versions. Project A needs `pandas 1.3`, Project B needs `pandas 2.0`. Without isolation they conflict.

A **virtual environment** is a self-contained directory with its own Python interpreter and packages, isolated from everything else on your system.

```
base environment        ds-notes environment
──────────────────      ────────────────────
Python 3.10             Python 3.10
pandas 2.0              pandas 1.5.3
numpy 1.24              numpy 1.23
                        scikit-learn 1.2
```

### Creating and using a conda environment

```bash
# Create a new environment named 'ds-notes' with Python 3.10
conda create --name ds-notes python=3.10

# Activate it
conda activate ds-notes

# Install packages
conda install pandas numpy matplotlib seaborn scikit-learn jupyter

# Deactivate when done
conda deactivate
```

### Common conda commands

| Command | Effect |
|---------|--------|
| `conda create --name myenv python=3.10` | Create new environment |
| `conda activate myenv` | Switch to environment |
| `conda deactivate` | Return to base |
| `conda env list` | List all environments |
| `conda install package` | Install a package |
| `conda list` | List installed packages |
| `conda remove --name myenv --all` | Delete an environment |

## Step 4 — Jupyter Notebook

Jupyter Notebook is an interactive environment where code, output, and text (Markdown) coexist in a single document.

```bash
# With your environment active:
jupyter notebook
# Opens browser at http://localhost:8888
```

### Key keyboard shortcuts

| Shortcut | Mode | Action |
|----------|------|--------|
| `Enter` | Command | Enter edit mode |
| `Esc` | Edit | Enter command mode |
| `Shift+Enter` | Either | Run cell, move to next |
| `Ctrl+Enter` | Either | Run cell, stay |
| `A` | Command | Insert cell above |
| `B` | Command | Insert cell below |
| `DD` | Command | Delete cell |
| `M` | Command | Change to Markdown cell |
| `Y` | Command | Change to code cell |
| `Ctrl+S` | Either | Save notebook |

### Markdown basics

```markdown
# Heading 1
## Heading 2
**bold**, *italic*, `inline code`
- bullet item
1. numbered item
```
