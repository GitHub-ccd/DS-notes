## What Problems Can Data Science Solve?

Data science sits at the intersection of statistics, computer science, and domain expertise. At its core, it uses data to answer questions and drive decisions. The major problem categories are:

### 1. Classification
Assign an item to a category.
- Is this email spam?
- Does this tumor image show cancer?
- Will this customer churn in the next 30 days?

### 2. Regression
Predict a continuous numerical value.
- What will this house sell for?
- How many units will we sell next quarter?
- What is the expected lifetime value of this customer?

### 3. Clustering
Discover natural groupings in unlabeled data.
- What customer segments exist in our user base?
- Which genes behave similarly across experiments?

### 4. Recommendation
Predict what a user will like based on past behavior.
- What movie should Netflix suggest next?
- What products are frequently bought together?

### 5. Natural Language Processing
Extract meaning from text or speech.
- What is the sentiment of this review (positive/negative)?
- What is this customer support email asking for?

### 6. Anomaly Detection
Flag unusual observations.
- Is this credit card transaction fraudulent?
- Is this server's behavior indicative of a security breach?

## The Data Science Process — OSEMN

Real data science projects follow a structured lifecycle. One common framework is **OSEMN** (pronounced "awesome"):

| Step | Name | What happens |
|------|------|--------------|
| **O** | Obtain | Collect data from databases, APIs, web scraping, sensors, spreadsheets |
| **S** | Scrub | Clean data — handle missing values, outliers, format inconsistencies |
| **E** | Explore | EDA — visualize distributions, find patterns, generate hypotheses |
| **M** | Model | Select, train, and evaluate a statistical or ML model |
| **N** | iNterpret | Communicate results, explain predictions, make decisions |

This process is **iterative**, not linear. Insights from exploration often require going back to scrubbing. Model results reveal data quality issues. Interpretation surfaces new questions.

```
Obtain → Scrub → Explore → Model → Interpret
           ↑___________________|         |
           |_____________________________|   ← feedback loops
```

## Types of Analytics

| Type | Question answered | Example |
|------|------------------|---------|
| **Descriptive** | What happened? | Monthly sales report |
| **Diagnostic** | Why did it happen? | Why did sales drop in Q3? |
| **Predictive** | What will happen? | Revenue forecast for Q4 |
| **Prescriptive** | What should we do? | Optimize ad spend allocation |

Most of this curriculum covers **predictive analytics** (ML models), but EDA is fundamentally **descriptive**, and understanding *why* requires **diagnostic** thinking.

## What Makes a Good Data Scientist?

Data scientists are effective when they can:
1. **Frame business problems** as data questions
2. **Work with messy, real-world data** without panic
3. **Communicate uncertainty** — knowing what a model can and cannot tell you
4. **Explain results to non-technical stakeholders**

The tools change constantly. The thinking process is durable.
