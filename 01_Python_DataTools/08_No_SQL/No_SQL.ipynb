# Section 8 — NoSQL Databases

NoSQL covers a broad family of database types that don't follow the relational (table/row) model. For ML practitioners in 2026, the two most relevant categories are **document stores** (MongoDB) and **vector databases** (ChromaDB, FAISS, Pinecone).

## 3. Vector Databases ← 2026 Priority

Vector databases store and search **high-dimensional embeddings** — the numerical representations that LLMs, image models, and recommendation systems produce. They are the backbone of RAG (Retrieval-Augmented Generation) pipelines and semantic search.

Key tools:
| Tool | Type | Notes |
|------|------|-------|
| **ChromaDB** | Local / hosted | Easiest to start with, great for notebooks |
| **FAISS** | Local library | Meta's highly optimised similarity search |
| **Pinecone** | Managed cloud | Production-grade, serverless |
| **Weaviate** | Self-hosted / cloud | Multimodal support |
