## Git Resources

### Official & Foundational
- [Pro Git Book (free)](https://git-scm.com/book/en/v2) — the authoritative reference; covers everything
  - [Getting Started — History of Git](https://git-scm.com/book/en/v2/Getting-Started-A-Short-History-of-Git)
  - [Getting Started — About Version Control](https://git-scm.com/book/en/v2/Getting-Started-About-Version-Control)
  - [Git Basics — Recording Changes](https://git-scm.com/book/en/v2/Git-Basics-Recording-Changes-to-the-Repository)
  - [Viewing the Commit History](https://git-scm.com/book/en/v2/Git-Basics-Viewing-the-Commit-History)
  - [Working with Remotes](https://git-scm.com/book/en/v2/Git-Basics-Working-with-Remotes)
  - [Branching — What a Branch Is](https://git-scm.com/book/en/v2/Git-Branching-Branches-in-a-Nutshell)
  - [Basic Branching and Merging](https://git-scm.com/book/en/v2/Git-Branching-Basic-Branching-and-Merging)
  - [Branch Management](https://git-scm.com/book/en/v2/Git-Branching-Branch-Management)
  - [Branching Workflows](https://git-scm.com/book/en/v2/Git-Branching-Branching-Workflows)
- [GitHub Git Handbook](https://guides.github.com/introduction/git-handbook/)

### Visual & Interactive
- [Learn Git Branching](https://learngitbranching.js.org/) — best interactive branching visualizer
- [Visual Git Guide](https://marklodato.github.io/visual-git-guide/index-en.html) — clear diagrams of core operations
- [Git Guide](https://rogerdudler.github.io/git-guide/) — concise cheat-sheet style overview

### Video

## Markdown Resources
- [GitHub Flavored Markdown Spec](https://github.github.com/github-flavored-markdown)
- [GitHub Docs — Writing on GitHub](https://docs.github.com/en/get-started/writing-on-github/getting-started-with-writing-and-formatting-on-github)

## Summary

A curated reference list for Git and Markdown. Note: Markdown is the text format used to display READMEs on GitHub, which is why README files use the `.md` extension.
