## 3. Dynamic Pages — Modern Tooling (2026)

> **Why this matters:** Many modern websites render content via JavaScript *after* the initial page load. `requests` + BeautifulSoup will return empty divs. You need a browser automation tool.

### Selenium → Playwright (upgrade)

**Selenium** was the standard for years but is slow, brittle, and requires manual WebDriver management. **Playwright** (by Microsoft) replaced it as the modern standard:

| | Selenium | Playwright |
|--|----------|------------|
| Speed | Slow | Fast (async-native) |
| Setup | WebDriver per browser | `playwright install` |
| API | Verbose | Clean, modern |
| Async | Bolted on | Native |
| Status (2026) | Legacy | Active, preferred |

```bash
pip install playwright
playwright install chromium
```

```python
from playwright.sync_api import sync_playwright

with sync_playwright() as p:
    browser = p.chromium.launch(headless=True)
    page = browser.new_page()
    page.goto('https://example.com')
    page.wait_for_selector('.data-table')   # wait for JS to render
    html = page.content()
    browser.close()

# Then parse html with BeautifulSoup as normal
from bs4 import BeautifulSoup
soup = BeautifulSoup(html, 'html.parser')
```

### When to use what

| Scenario | Tool |
|----------|------|
| Static HTML | `requests` + `BeautifulSoup` |
| JavaScript-rendered content | `Playwright` |
| API available | Use the API instead (always prefer) |
| Large-scale crawling | `Scrapy` |

## 4. Ethics & robots.txt

Always check `https://example.com/robots.txt` before scraping. Respect `Disallow` directives. Add delays between requests (`time.sleep(1)`). Do not scrape data that is available via an official API.
