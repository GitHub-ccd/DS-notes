## HTML — Structure and Meaning

HTML (HyperText Markup Language) describes the **structure** of a web page using **tags**. Tags wrap content and tell the browser how to interpret it. They don't contain logic — they just label things.

```html
<p>This is a paragraph.</p>
```

The opening tag is `<p>`, the closing tag is `</p>`, and the content sits between them. Tags can have **attributes** inside the opening tag:

```html
<a href="http://www.example.com">Click here</a>
```

`href` is the attribute name, the URL is its value. When scraping, attributes are often where the useful data lives.

### Document structure

Every HTML page has the same skeleton:

```html
<!DOCTYPE html>
<html>
    <head>
        <!-- metadata: title, stylesheet links, scripts -->
        <title>Page Title</title>
    </head>
    <body>
        <!-- visible content goes here -->
    </body>
</html>
```

The `<head>` holds metadata; the `<body>` holds everything you see on screen. When scraping, you almost always work inside `<body>`.

## Common HTML Elements

### Headings

Six levels, `h1` (largest) to `h6` (smallest). Search engines use heading hierarchy to understand page content — scrapers use them as landmarks.

```html
<h1>Main Title</h1>
<h2>Section Title</h2>
<h3>Subsection</h3>
```

### Paragraphs and links

```html
<p>Some text with a <a href="http://example.com">link</a> inside it.</p>
```

The `<a>` tag's `href` attribute is the URL. When scraping a list of links, you extract `a` tags and read their `href`.

### Images

```html
<img src="photo.jpg" alt="Description of the image">
```

No closing tag. The `src` attribute is the image URL. When scraping images, you extract `img` tags and read their `src`.

### Lists

```html
<!-- Unordered (bullet) list -->
<ul>
    <li>First item</li>
    <li>Second item</li>
</ul>

<!-- Ordered (numbered) list -->
<ol>
    <li>Step one</li>
    <li>Step two</li>
</ol>
```

### Divs and spans — containers

`<div>` is a generic block container; `<span>` is a generic inline container. They carry no inherent meaning but are used heavily to group and style content. Most of what you'll target when scraping is inside a `<div>` somewhere.

```html
<div class="product-card">
    <h3>Book Title</h3>
    <p class="price">£12.99</p>
</div>
```

### `class` and `id` attributes

These two attributes are the primary way you'll **target** elements when scraping:

- `class` — can appear on multiple elements; used for groups of similar things (e.g. all product cards)
- `id` — should be unique per page; used to identify one specific element

```html
<p class="price_color">£12.99</p>      <!-- class selector: .price_color -->
<div id="main-content">...</div>        <!-- id selector: #main-content -->
```

## CSS — Appearance

CSS (Cascading Style Sheets) controls how HTML elements look. HTML handles *what* is on the page; CSS handles *how it looks*. The same HTML page can look completely different depending on the CSS applied to it — this is the idea behind [CSS Zen Garden](http://www.csszengarden.com/):

![CSS Zen Garden 1](assets/intro_to_css/zengarden1.png)
![CSS Zen Garden 2](assets/intro_to_css/zengarden2.png)

All the differences above are purely CSS — the underlying HTML is identical.

As a scraper you don't *write* CSS, but you use CSS selector syntax to *target* elements — Beautiful Soup uses it too.

---

## CSS Selectors

A CSS **selector** specifies which HTML elements a rule applies to.

| Selector | Syntax | Targets |
|----------|--------|---------|
| Type | `p` | All `<p>` elements |
| Class | `.price_color` | All elements with `class="price_color"` |
| ID | `#main-header` | The element with `id="main-header"` |
| Descendant | `div p` | All `<p>` elements inside a `<div>` |

### CSS declaration blocks

A **declaration block** pairs a selector with one or more `property: value` rules:

```css
/* Select all h1 tags */
h1 {
    color: blue;
    font-family: Georgia;
}

/* Select elements with class 'price_color' */
.price_color {
    color: green;
    font-weight: bold;
}

/* Select the element with id 'main-header' */
#main-header {
    font-size: 22px;
}
```

Every declaration ends with a semicolon. The selector is the key — it's exactly what you pass to `soup.find_all('p', class_='price_color')` in Beautiful Soup.

## Reading a Real Page's Source

In your browser, right-click any element on a page and choose **Inspect** (or **Inspect Element**). The DevTools panel shows the live HTML tree. Things to note:

- Tags are nested — a `<div>` may contain many `<p>` tags which may contain `<a>` tags
- Every visible piece of data lives inside some tag with some class or id
- Hover over a tag in DevTools to highlight it on the page

The workflow when scraping is:
1. Find the element you want in DevTools
2. Note its tag name and its `class` or `id`
3. Use that selector in Beautiful Soup
