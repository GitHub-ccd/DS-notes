# Beautiful Soup Basics

## Introduction

Beautiful Soup is a Python library for parsing HTML documents and navigating their structure. It turns an HTML string into a tree of objects you can query with selectors — the same selectors you just learned from CSS. This notebook covers the core API: loading a page, finding elements, reading attributes, and navigating the tree.

## Objectives

You will be able to:

- Parse an HTML document with `BeautifulSoup`
- Select elements by tag name, class, and id using `find` and `find_all`
- Read tag attributes with `.attrs`
- Navigate the tree using parent and sibling relationships

---

## The Document Object Model (DOM)

A web page is more than raw HTML text — the browser parses it into a tree of objects called the **Document Object Model (DOM)**. Each tag becomes a node; nodes have parents, children, and siblings.

<img src="assets/web_scraping_with_beautiful_soup/DOM-model.svg.png" width="500">

Beautiful Soup builds the same tree structure in Python, letting you navigate it just like you'd navigate a family tree: go up to a parent, across to a sibling, or down to a child.

---

## Parsing HTML with Beautiful Soup

````python
from bs4 import BeautifulSoup

# Parse from a local file
with open('assets/web_scraping_with_beautiful_soup/sample_page.html') as f:
    soup = BeautifulSoup(f, 'html.parser')

# prettify() prints the tree with indentation
print(soup.prettify())
````

The result is a `BeautifulSoup` object — a navigable tree. You can also parse a string directly:

```python
html_string = '<p class="title"><b>Hello</b></p>'
soup = BeautifulSoup(html_string, 'html.parser')
```

Or parse a live web page (covered in the next notebook):

```python
import requests
response = requests.get('http://example.com')
soup = BeautifulSoup(response.content, 'html.parser')
```

---

## Finding Elements

### `find()` — first match

`soup.find(tag)` returns the **first** element with that tag name.

````python
# First <p> tag
print(soup.find('p'))

# Shorthand dot notation — same result
print(soup.p)
````

````python
# Find by id
soup.find(id='link1')
````

````python
# Find by tag + class
soup.find('p', class_='title')
````

### `find_all()` — all matches

`find_all()` returns a **list** of all matching elements.

````python
# All <a> tags on the page
links = soup.find_all('a')
print(f'{len(links)} links found')
for link in links:
    print(link)
````

---

## Reading Tag Content and Attributes

````python
tag = soup.find('a')  # first <a> tag

print('tag name:    ', tag.name)
print('text content:', tag.string)
print('all attrs:   ', tag.attrs)
print('href attr:   ', tag.attrs['href'])
print('class attr:  ', tag.attrs['class'])
````

````python
# Extract all hrefs from all links
hrefs = [a.attrs['href'] for a in soup.find_all('a')]
hrefs
````

---

## Tree Navigation — Parents, Children, Siblings

````python
# Parent of the <title> tag
print(soup.title.parent.name)  # 'head'
````

````python
# Children of <body>
for child in soup.body.children:
    if child.name:  # skip whitespace NavigableString nodes
        print(child.name, '->', child.attrs)
````

````python
# Next sibling — move across the tree at the same level
first_p = soup.find('p')
print('first p:', first_p)

# nextSibling can return whitespace text nodes; keep calling until you get a tag
sibling = first_p.nextSibling
while sibling and not sibling.name:
    sibling = sibling.nextSibling
print('next p:', sibling)
````

## Ethics and Precautions

Web scraping is powerful but comes with responsibilities:

- **Rate limiting** — making hundreds of requests per second can overload a server and may be considered a denial-of-service attack. Add delays with `time.sleep()`.
- **Terms of service** — many websites explicitly prohibit scraping. Check `robots.txt` and the site's ToS before scraping.
- **Login-gated content** — scraping pages that require login may expose you to legal liability.
- **IP blocking** — aggressive scraping will get your IP blacklisted. Test on small samples first.

### When BS4 isn't enough

Beautiful Soup only parses static HTML. Many modern sites load content dynamically with JavaScript — the initial HTML is nearly empty, and data appears only after JS runs. For those sites you need:

- **Selenium** — controls a real browser, waits for JS to execute
- **Playwright** — faster modern alternative to Selenium
- **Scrapy** — full crawling framework for large-scale scraping
