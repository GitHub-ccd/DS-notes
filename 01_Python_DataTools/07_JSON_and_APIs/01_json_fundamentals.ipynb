# JSON Fundamentals

## Introduction

JSON (JavaScript Object Notation) is the standard data format for the web. When you retrieve data from an API, read a config file, or exchange data between services, it will almost certainly arrive as JSON. This notebook covers the Python `json` module — how to load, navigate, and export JSON data — and how to convert it to a pandas DataFrame.

## Loading a JSON File

Open the file with Python's built-in `open()` and pass the file object to `json.load()`. The result is a native Python object — usually a `dict` at the root level.

```python
import json
import pandas as pd

with open('data/json/nyc_2001_campaign_finance.json') as f:
    data = json.load(f)

print(type(data))
```

```python
# Top-level keys
data.keys()
```

```python
# Check what type each value is
for key, value in data.items():
    print(key, '->', type(value))
```

---

## Navigating Nested JSON

JSON files are often deeply nested. Standard dictionary and list indexing gets you to any level.

```python
# The 'meta' key holds metadata — view it as a DataFrame
pd.DataFrame.from_dict(data['meta'])
```

```python
# The 'data' key is the actual records
print(type(data['data']), '->', len(data['data']), 'rows')

# Preview the first record
data['data'][0]
```

```python
# Column names are stored in the metadata
# Navigate: meta -> view -> columns -> list of column dicts
col_defs = data['meta']['view']['columns']
print(type(col_defs), len(col_defs), 'columns')

col_names = [col['name'] for col in col_defs]
print(col_names)
```

---

## Converting JSON to a DataFrame

Once you've located the data list and the column names, building a DataFrame is straightforward.

```python
df = pd.DataFrame(data['data'], columns=col_names)
print(df.shape)
df.head(3)
```

```python
# The first row is a header row baked into the data — drop it
df = df.iloc[1:].reset_index(drop=True)
df.head(3)
```

---

## Saving Back to JSON

`json.dump()` writes a Python object back to a JSON file.

```python
# Save with indent for human-readable output
with open('data/json/output.json', 'w') as f:
    json.dump(data, f, indent=2)

print('Saved.')
```
