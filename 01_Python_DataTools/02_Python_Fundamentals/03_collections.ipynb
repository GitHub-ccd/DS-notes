## Lists

```python
# Creating lists
temperatures = [72, 68, 74, 70, 65, 71, 73]
mixed = [1, 'hello', True, None, [2, 3]]  # any types allowed
empty = []

# Indexing and slicing (same as strings)
print(temperatures[0])    # 72 — first element
print(temperatures[-1])   # 73 — last element
print(temperatures[2:5])  # [74, 70, 65] — slice

# Length
print(len(temperatures))  # 7
```

```python
# List methods — lists ARE mutable (unlike strings)
fruits = ['apple', 'banana', 'cherry']

fruits.append('date')         # add to end
fruits.insert(1, 'blueberry') # insert at index
fruits.extend(['elderberry', 'fig'])  # add multiple
print(fruits)

fruits.remove('banana')    # remove first occurrence
popped = fruits.pop()      # remove and return last item
popped_idx = fruits.pop(0) # remove and return item at index
print(f"Popped: {popped}, {popped_idx}")

# Sorting
nums = [3, 1, 4, 1, 5, 9, 2, 6]
nums.sort()                  # sorts in place
sorted_copy = sorted(nums)   # returns new list, original unchanged
nums.reverse()               # reverses in place

# Searching
print(3 in nums)             # True
print(nums.index(5))         # index of value 5
print(nums.count(1))         # how many times 1 appears
```

### List Comprehensions

A concise, readable way to create lists — preferred over explicit for-loops for simple transformations.

```python
# Basic comprehension: [expression for item in iterable]
squares = [x**2 for x in range(1, 6)]
print(squares)  # [1, 4, 9, 16, 25]

# With a filter: [expression for item in iterable if condition]
even_squares = [x**2 for x in range(1, 11) if x % 2 == 0]
print(even_squares)  # [4, 16, 36, 64, 100]

# Transform strings
names = ['alice', 'bob', 'charlie']
capitalized = [n.capitalize() for n in names]
print(capitalized)

# Flatten nested list
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
flat = [x for row in matrix for x in row]
print(flat)
```

## Dictionaries

```python
# Dictionaries: key-value pairs, ordered (Python 3.7+)
person = {
    'name': 'Alice',
    'age': 30,
    'city': 'New York',
    'is_active': True
}

# Access values
print(person['name'])            # 'Alice'
print(person.get('email'))       # None — safer than [] (no KeyError)
print(person.get('email', ''))   # '' — default value if key missing

# Modify
person['age'] = 31               # update existing
person['email'] = 'a@mail.com'   # add new key
del person['is_active']          # delete key

# Check for key
print('name' in person)          # True
print('salary' in person)        # False
```

```python
# Iterating over dicts
person = {'name': 'Alice', 'age': 31, 'city': 'New York'}

for key in person:               # iterate keys
    print(key)

for val in person.values():      # iterate values
    print(val)

for key, val in person.items():  # iterate key-value pairs
    print(f"{key}: {val}")
```

```python
# Dict comprehension
words = ['apple', 'banana', 'cherry']
word_lengths = {w: len(w) for w in words}
print(word_lengths)  # {'apple': 5, 'banana': 6, 'cherry': 6}

# Invert a dictionary
original = {'a': 1, 'b': 2, 'c': 3}
inverted = {v: k for k, v in original.items()}
print(inverted)  # {1: 'a', 2: 'b', 3: 'c'}

# Useful dict methods
d = {'x': 1, 'y': 2}
d.update({'y': 99, 'z': 3})   # merge another dict (or keyword args)
print(d)

# setdefault — get value, or set it if missing
counts = {}
for char in 'mississippi':
    counts[char] = counts.get(char, 0) + 1
print(counts)
```

## Nested Data Structures

Real-world data (JSON from APIs, database rows) is often nested.

```python
# List of dicts — a common pattern (like a database table)
students = [
    {'name': 'Alice', 'grade': 'A', 'score': 95},
    {'name': 'Bob',   'grade': 'B', 'score': 82},
    {'name': 'Carol', 'grade': 'A', 'score': 91},
]

# Access specific values
print(students[0]['name'])   # 'Alice'
print(students[-1]['score']) # 91

# Extract all names with list comprehension
names = [s['name'] for s in students]
print(names)

# Filter: only A students
a_students = [s for s in students if s['grade'] == 'A']
print(a_students)

# Average score
avg_score = sum(s['score'] for s in students) / len(students)
print(f"Average: {avg_score:.1f}")
```

## Other Collection Types

| Type | Ordered | Mutable | Duplicates | Use case |
|------|---------|---------|------------|----------|
| `list` | Yes | Yes | Yes | Sequence, stack, queue |
| `tuple` | Yes | No | Yes | Fixed record, dict key |
| `set` | No | Yes | No | Membership testing, deduplication |
| `dict` | Yes (3.7+) | Yes | Keys: no | Key-value lookup |

```python
# tuple — immutable list
coords = (40.7128, -74.0060)  # lat, lon — shouldn't change

# set — unordered, no duplicates
raw_ids = [1, 2, 2, 3, 3, 3, 4]
unique_ids = set(raw_ids)
print(unique_ids)   # {1, 2, 3, 4}

# Set operations
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}
print(a | b)  # union:        {1, 2, 3, 4, 5, 6}
print(a & b)  # intersection: {3, 4}
print(a - b)  # difference:   {1, 2}
```

## Summary

- **Lists** are ordered, mutable sequences — use for collections where order matters and items can change
- **Dicts** are key-value stores — use for named attributes and fast lookups by key
- **List comprehensions** `[expr for x in iterable if cond]` are clean and Pythonic — prefer them over explicit for-loops for simple transformations
- **Nested structures** (lists of dicts) model tabular data before you have Pandas
- Use **tuples** for immutable records, **sets** for deduplication and membership tests
