## Creating Strings

```python
# Single quotes, double quotes — interchangeable
name = 'Alice'
city = "New York"

# Use the other quote type to avoid escaping
msg = "It's a beautiful day"   # no escape needed
msg = 'She said "hello"'        # no escape needed

# Escape sequences
tab = "col1\tcol2"     # \t = tab
newline = "line1\nline2"  # \n = newline
path = "C:\\Users\\name"  # \\ = literal backslash
raw = r"C:\Users\name"   # r"" = raw string, backslashes are literal

# Triple quotes — multiline strings
docstring = """
This is a
multi-line string.
"""
print(repr(docstring))
```

## Indexing and Slicing

Strings are sequences — you can access individual characters by index and extract substrings with slices.

```python
s = "Hello, World!"
#    0123456789...
#   -13...       -1  (negative indices count from end)

print(s[0])      # 'H'  — first character
print(s[-1])     # '!'  — last character
print(s[7:12])   # 'World'  — slice [start:stop] (stop is exclusive)
print(s[:5])     # 'Hello'  — from start to index 5
print(s[7:])     # 'World!' — from index 7 to end
print(s[::2])    # every other character (step=2)
print(s[::-1])   # reversed string

print(len(s))    # 13 — number of characters
```

## String Methods

Strings are **immutable** — methods return new strings, they never modify in place.

```python
text = "  Hello, World!  "

# Case
print(text.upper())       # '  HELLO, WORLD!  '
print(text.lower())       # '  hello, world!  '
print(text.title())       # '  Hello, World!  '

# Whitespace
print(text.strip())       # 'Hello, World!'   — removes both ends
print(text.lstrip())      # 'Hello, World!  ' — left only
print(text.rstrip())      # '  Hello, World!' — right only

# Search
clean = text.strip()
print(clean.find('World'))    # 7   — index of first occurrence (-1 if not found)
print(clean.count('l'))       # 3   — how many times 'l' appears
print('World' in clean)       # True
print(clean.startswith('H'))  # True
print(clean.endswith('!'))    # True

# Replace and remove
print(clean.replace('World', 'Python'))  # 'Hello, Python!'
print(clean.replace('l', ''))            # 'Heo, Word!'

# Split and join
words = "one,two,three".split(',')   # ['one', 'two', 'three']
print(words)
print(' | '.join(words))              # 'one | two | three'
```

```python
# Useful for data cleaning
column_name = "  Total Revenue (USD)  "
clean_col = column_name.strip().lower().replace(' ', '_').replace('(', '').replace(')', '')
print(clean_col)  # 'total_revenue_usd'
```

## String Formatting

Three ways to insert values into strings — f-strings are the modern standard.

```python
name = "Alice"
score = 0.9567
count = 1_234_567

# f-strings (Python 3.6+) — preferred
print(f"Name: {name}")
print(f"Score: {score:.2f}")        # 2 decimal places
print(f"Score: {score:.1%}")        # percentage format
print(f"Count: {count:,}")          # thousands separator
print(f"Count: {count:>12,}")       # right-aligned in 12-char field

# Expression inside f-strings
print(f"2 + 2 = {2 + 2}")
print(f"Upper: {name.upper()}")

# .format() — older but still common
print("Name: {}, Score: {:.2f}".format(name, score))
print("Name: {name}, Score: {score:.2f}".format(name=name, score=score))

# % formatting — legacy, avoid in new code
print("Name: %s, Score: %.2f" % (name, score))
```

## Concatenation

```python
# + operator — creates a new string
greeting = "Hello" + ", " + "World!"
print(greeting)

# Repetition
print("-" * 40)
print("ha" * 3)   # 'hahaha'

# Building strings in a loop — use join(), not +=
words = ['data', 'science', 'is', 'fun']

# Bad (O(n²) — creates a new string each iteration):
result = ""
for w in words:
    result += w + " "

# Good (O(n) — collect then join once):
result = " ".join(words)
print(result)
```

## Summary

Key string concepts:
- Strings are **immutable** — methods return new strings
- Use **f-strings** for formatting — `f"value: {x:.2f}"`
- `.strip()`, `.lower()`, `.replace()` are your primary data-cleaning tools
- Use `" ".join(list)` not `+=` inside loops for performance
- Slicing: `s[start:stop:step]` — stop is exclusive, negative indices count from end
