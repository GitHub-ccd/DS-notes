# Conditionals

## Introduction

Conditional statements allow programs to make decisions — executing different code depending on whether a condition is true or false. This is the core of all logic and branching in your code.

## Objectives

You will be able to:

* Write `if`, `elif`, and `else` blocks
* Use comparison and logical operators
* Apply truthiness for concise conditions
* Write conditional expressions (ternary)
* Avoid common conditional pitfalls

---

## Basic if/elif/else

```python
temperature = 85

if temperature > 90:
    print("Very hot")
elif temperature > 75:
    print("Warm")       # this executes
elif temperature > 60:
    print("Mild")
else:
    print("Cold")

# Notes:
# - Colon : is required at end of each condition
# - Body must be indented (4 spaces)
# - elif = else if (shortcut, avoids nesting)
# - Only ONE branch executes — first True condition wins
# - else is a catch-all, no condition needed
```

## Comparison Operators

```python
x, y = 5, 10

print(x == y)   # False  — equality (NOT assignment!)
print(x != y)   # True   — not equal
print(x < y)    # True   — less than
print(x <= y)   # True   — less than or equal
print(x > y)    # False
print(x >= y)   # False

# Chained comparisons — very readable in Python
age = 25
print(18 <= age < 65)   # True — between 18 and 64 inclusive

# Identity vs equality
a = [1, 2, 3]
b = [1, 2, 3]
print(a == b)    # True  — same content
print(a is b)    # False — different objects in memory

# Use 'is' only for None, True, False checks:
result = None
print(result is None)  # True — correct
print(result == None)  # True — works, but is not idiomatic
```

## Logical Operators

```python
age = 25
has_license = True

# and — both conditions must be True
if age >= 16 and has_license:
    print("Can drive")

# or — at least one must be True
is_admin = False
is_owner = True
if is_admin or is_owner:
    print("Has elevated access")

# not — inverts the boolean
is_empty = []
if not is_empty:
    print("List is empty")

# Short-circuit evaluation — Python stops evaluating as soon as result is known
# In 'and': if first is False, second is never evaluated
# In 'or':  if first is True, second is never evaluated
data = None
if data is not None and len(data) > 0:  # safe — len() only called if data is not None
    print("Has data")
```

---

## Truthiness in Conditionals

```python
# Verbose but redundant:
my_list = [1, 2, 3]
if len(my_list) > 0:
    print("Not empty")

# Pythonic — rely on truthiness:
if my_list:
    print("Not empty")

# Falsy values: 0, 0.0, False, None, "", [], {}, set()
# Everything else is truthy

user = None
if user:
    print(user['name'])
else:
    print("No user")
```

## Conditional Expressions (Ternary)

For simple two-way choices, Python has a one-line conditional expression.

```python
# Syntax: value_if_true if condition else value_if_false

score = 75
grade = 'Pass' if score >= 60 else 'Fail'
print(grade)  # 'Pass'

# Assign a default for None values
user_input = None
name = user_input if user_input is not None else 'Anonymous'
print(name)

# Equivalent to (but more concise than):
if user_input is not None:
    name = user_input
else:
    name = 'Anonymous'

# Don't nest ternaries — it becomes unreadable:
# a if x else b if y else c   ← avoid
```

## Common Patterns in Data Science

```python
# Categorizing numeric values
def categorize_bmi(bmi):
    if bmi < 18.5:
        return 'Underweight'
    elif bmi < 25.0:
        return 'Normal'
    elif bmi < 30.0:
        return 'Overweight'
    else:
        return 'Obese'

print(categorize_bmi(22.3))  # 'Normal'

# Guard clause — return early to avoid deep nesting
def process_record(record):
    if record is None:
        return None
    if 'value' not in record:
        return None
    # main logic here, no indentation hell
    return record['value'] * 2
```

## Summary

- `if/elif/else` — only the first matching branch executes
- Use `and`, `or`, `not` to combine conditions; rely on short-circuit evaluation for safety
- Exploit truthiness: `if my_list:` instead of `if len(my_list) > 0:`
- Use `is` for None/True/False, `==` for value comparison
- Ternary: `value_if_true if condition else value_if_false`
- Guard clauses (early returns) keep functions flat and readable
