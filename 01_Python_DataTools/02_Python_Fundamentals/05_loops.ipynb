# Loops

## Introduction

Loops let you repeat code automatically — essential for processing every row in a dataset, computing statistics over arrays, or running iterative algorithms. Python has two loop types: `for` (iterate over a sequence) and `while` (repeat while a condition holds).

## Objectives

You will be able to:

* Write `for` loops over lists, strings, dicts, and ranges
* Use `enumerate()`, `zip()`, and `range()` effectively
* Write `while` loops with proper exit conditions
* Control loop flow with `break`, `continue`, and `else`
* Write nested loops and understand their complexity

## for Loops

```python
# Iterate over a list
fruits = ['apple', 'banana', 'cherry']
for fruit in fruits:
    print(fruit)

# Iterate over a string (character by character)
for char in 'hello':
    print(char, end=' ')  # end=' ' suppresses newline
print()  # newline at end

# Iterate over a dict
scores = {'Alice': 95, 'Bob': 82, 'Carol': 91}
for name, score in scores.items():
    print(f"{name}: {score}")
```

```python
# range() — generate a sequence of numbers
for i in range(5):         # 0, 1, 2, 3, 4
    print(i, end=' ')
print()

for i in range(1, 6):      # 1, 2, 3, 4, 5 (stop is exclusive)
    print(i, end=' ')
print()

for i in range(0, 10, 2):  # 0, 2, 4, 6, 8 (step=2)
    print(i, end=' ')
print()

for i in range(5, 0, -1):  # 5, 4, 3, 2, 1 (countdown)
    print(i, end=' ')
print()
```

```python
# enumerate() — get index AND value simultaneously
teams = ['Alpha', 'Beta', 'Gamma']
for i, team in enumerate(teams):
    print(f"{i+1}. {team}")

# start parameter
for i, team in enumerate(teams, start=1):
    print(f"{i}. {team}")

# zip() — iterate over two sequences in parallel
names = ['Alice', 'Bob', 'Carol']
scores = [95, 82, 91]
for name, score in zip(names, scores):
    print(f"{name}: {score}")

# zip stops at the shortest sequence
```

## while Loops

```python
# Repeat while condition is True
count = 0
while count < 5:
    print(count)
    count += 1   # ALWAYS update the condition variable — infinite loop risk!

# while is natural for: convergence loops, reading until EOF, waiting for input
# for is natural for: iterating a known collection or count
```

```python
# Classic while pattern: gradient descent (simplified)
x = 10.0
learning_rate = 0.1
tolerance = 0.001
iterations = 0

while abs(x) > tolerance:
    x = x - learning_rate * x  # gradient of x^2 is 2x, simplified here
    iterations += 1

print(f"Converged to x={x:.6f} after {iterations} iterations")
```

## break, continue, and else

```python
# break — exit the loop immediately
for num in [1, 5, 3, 8, 2, 7]:
    if num > 6:
        print(f"Found first value > 6: {num}")
        break

# continue — skip to next iteration
for num in range(10):
    if num % 2 == 0:
        continue  # skip even numbers
    print(num, end=' ')  # prints: 1 3 5 7 9
print()

# else clause — runs if loop completes without break
# Useful for "search" loops
target = 42
data = [10, 25, 17, 38, 55]
for val in data:
    if val == target:
        print(f"Found {target}")
        break
else:
    print(f"{target} not found")  # this runs
```

## Nested Loops

```python
# Multiplication table
for i in range(1, 4):
    for j in range(1, 4):
        print(f"{i*j:3}", end='')
    print()  # newline after each row

print()

# Nested loops over data
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
row_sums = []
for row in matrix:
    total = 0
    for val in row:
        total += val
    row_sums.append(total)
print(row_sums)  # [6, 15, 24]

# List comprehension version:
row_sums = [sum(row) for row in matrix]
print(row_sums)
```

```python
# Complexity warning: nested loops are O(n*m)
# A 3-level nested loop over n items is O(n^3) — expensive!
# Always ask: can this be vectorized with NumPy instead?

# Instead of:
data = list(range(1000))
total = 0
for x in data:
    total += x

# Use:
import numpy as np
total = np.sum(data)  # ~100x faster for large arrays
print(total)
```

## Summary

| Loop type | Use when |
|-----------|----------|
| `for x in collection` | You know what to iterate over |
| `for i in range(n)` | You need index-based iteration |
| `for i, x in enumerate(...)` | You need both index and value |
| `for a, b in zip(...)` | You need to iterate two sequences together |
| `while condition` | Repeat until something converges or changes |

Key control flow: `break` exits the loop, `continue` skips to next iteration, `else` runs if no `break` occurred.

For performance on large data: prefer NumPy vectorized operations over explicit Python loops.
