# Variables and Data Types

## Introduction

Variables are named containers for data. Python is **dynamically typed** — you don't declare the type of a variable; Python infers it at runtime. This section covers the basic scalar types: integers, floats, booleans, and `None`.

## Objectives

You will be able to:

* Assign and reassign variables
* Distinguish between int, float, bool, and NoneType
* Use type coercion and type checking functions
* Explain Python's naming rules and conventions

---

## Assignment

```python
# Basic assignment — single value
temperature = 98.6
city = "New York"
is_raining = False

# Multiple assignment
x, y, z = 1, 2, 3

# Same value to multiple variables
a = b = c = 0

# Reassignment — Python just rebinds the name
count = 10
count = count + 1  # or: count += 1
print(count)  # 11
```

---

## Integers and Floats

```python
# int — whole numbers (no size limit in Python 3)
population = 8_000_000_000    # underscores for readability
negative = -42

# float — decimal numbers (64-bit IEEE 754 double)
pi = 3.14159
scientific = 1.5e10           # 1.5 × 10^10

# Arithmetic operators
print(10 + 3)   # 13  addition
print(10 - 3)   # 7   subtraction
print(10 * 3)   # 30  multiplication
print(10 / 3)   # 3.333...  true division (always returns float)
print(10 // 3)  # 3   floor division
print(10 % 3)   # 1   modulo (remainder)
print(10 ** 3)  # 1000  exponentiation
```

```python
# Floating point precision — a well-known gotcha
print(0.1 + 0.2)          # 0.30000000000000004 — not exactly 0.3
print(round(0.1 + 0.2, 2))  # 0.3 — use round() when precision matters

# For financial calculations, use the decimal module:
from decimal import Decimal
print(Decimal('0.1') + Decimal('0.2'))  # 0.3
```

---

## Booleans

```python
# Boolean values — True and False (capitalized)
is_valid = True
is_empty = False

# Boolean operators
print(True and False)   # False
print(True or False)    # True
print(not True)         # False

# Comparison operators return booleans
print(5 > 3)    # True
print(5 == 5)   # True  (== is equality, = is assignment)
print(5 != 4)   # True
print(5 >= 5)   # True

# Booleans are subclass of int
print(int(True))   # 1
print(int(False))  # 0
print(True + True)  # 2 — useful for counting truthy conditions
```

---

## Truthiness and NoneType

```python
# None — represents absence of a value (like null in other languages)
result = None
print(type(result))    # <class 'NoneType'>
print(result is None)  # True — use 'is' not '==' to check for None

# Truthiness — any object can be evaluated in a boolean context
# Falsy values:
falsy = [0, 0.0, False, None, "", [], {}, set()]
for val in falsy:
    print(f"{repr(val):15} → {bool(val)}")

# Everything else is truthy
print(bool(42))        # True
print(bool("hello"))   # True
print(bool([1, 2]))    # True
```

---

## Type Checking and Coercion

```python
# Check type
x = 42
print(type(x))          # <class 'int'>
print(isinstance(x, int))  # True  ← preferred in production code

# Type coercion — explicit conversion between types
print(int(3.7))    # 3     — truncates, does NOT round
print(float(5))    # 5.0
print(str(100))    # '100'
print(bool(1))     # True
print(int('42'))   # 42    — string to int works if the string is a number

# What fails:
try:
    int('hello')           # ValueError
except ValueError as e:
    print(f"Error: {e}")
```

## Naming Rules

| Rule | Examples |
|------|----------|
| Letters, digits, underscores only | `my_var`, `count2`, `_private` |
| Cannot start with a digit | ~~`2fast`~~ |
| Case-sensitive | `score` ≠ `Score` ≠ `SCORE` |
| Cannot use keywords | ~~`for`, `if`, `class`, `return`~~ |
| Convention: `snake_case` | `customer_count`, `avg_salary` |

## Summary

| Type | Example | Falsy value |
|------|---------|-------------|
| `int` | `42`, `-1`, `0` | `0` |
| `float` | `3.14`, `1e10` | `0.0` |
| `bool` | `True`, `False` | `False` |
| `NoneType` | `None` | `None` |

Key points:
- `/` always returns float; `//` returns floor division (int for int inputs)
- Use `isinstance()` not `type() ==` for type checking
- Use `is None` not `== None` for None checks
- Floating point arithmetic has precision limits — use `round()` or `Decimal` when it matters
