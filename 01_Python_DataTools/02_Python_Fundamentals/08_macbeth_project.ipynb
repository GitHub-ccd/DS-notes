## Step 1 — Get the Text

```python
import urllib.request

# Download Macbeth from Project Gutenberg
url = 'https://www.gutenberg.org/cache/epub/2264/pg2264.txt'
response = urllib.request.urlopen(url)
macbeth = response.read().decode('utf-8')

print(f"Total characters: {len(macbeth):,}")
print(f"First 500 characters:\n{macbeth[:500]}")
```

---

## Step 2 — Explore the Raw Text

```python
# Find where the actual play starts (after Project Gutenberg header)
start_idx = macbeth.find('THE TRAGEDY OF MACBETH')
print(f"Play starts at index: {start_idx}")

# Trim to just the play text
macbeth_play = macbeth[start_idx:]
print(f"Play length: {len(macbeth_play):,} characters")
print(f"Preview:\n{macbeth_play[:300]}")
```

---

## Step 3 — Tokenize (Split into Words)

```python
# Normalize to lowercase and split on whitespace
raw_words = macbeth_play.lower().split()
print(f"Raw word count: {len(raw_words):,}")
print(f"First 20 words: {raw_words[:20]}")
```

```python
import string

# Clean punctuation from each word
def clean_word(word):
    """Strip leading/trailing punctuation and return the clean word."""
    return word.strip(string.punctuation)

# Apply cleaning and remove empty strings
words = [clean_word(w) for w in raw_words if clean_word(w)]
print(f"Clean word count: {len(words):,}")
print(f"Unique words: {len(set(words)):,}")
```

---

## Step 4 — Build Word Frequency Dictionary

```python
# Count occurrences of each word
word_counts = {}
for word in words:
    word_counts[word] = word_counts.get(word, 0) + 1

print(f"Unique words tracked: {len(word_counts):,}")

# Look up specific words
for w in ['macbeth', 'king', 'murder', 'blood', 'witches']:
    print(f"  '{w}': {word_counts.get(w, 0)} occurrences")
```

---

## Step 5 — Top 25 Words

```python
# Sort by count (descending)
sorted_words = sorted(word_counts.items(), key=lambda item: item[1], reverse=True)

print("Top 25 words:")
for word, count in sorted_words[:25]:
    print(f"  {word:<20} {count:>5}")
```

```python
# The top words are mostly stop words (the, and, of...) — common but not meaningful
# Filter them out
stop_words = {
    'the', 'and', 'of', 'to', 'a', 'i', 'in', 'is', 'it',
    'that', 'with', 'he', 'as', 'his', 'be', 'this', 'are',
    'for', 'by', 'not', 'we', 'but', 'have', 'my', 'you',
    'from', 'or', 'an', 'at', 'our', 'me', 'so', 'your',
    'all', 'no', 'do', 'what', 'will', 'thee', 'thou', 'shall',
    'on', 'him', 'which', 'may', 'has', 'her', 'was', 'their',
    'had', 'there', 'more', 'if', 'them', 'come', 'then',
}

content_words = [(w, c) for w, c in sorted_words if w not in stop_words]

print("Top 25 content words (stop words removed):")
for word, count in content_words[:25]:
    print(f"  {word:<20} {count:>5}")
```

---

## Step 6 — Visualize

```python
import matplotlib.pyplot as plt
%matplotlib inline

# Top 25 content words as a horizontal bar chart
top_n = 25
labels = [w for w, c in content_words[:top_n]]
counts = [c for w, c in content_words[:top_n]]

fig, ax = plt.subplots(figsize=(10, 8))
ax.barh(labels[::-1], counts[::-1])  # flip so highest is at top
ax.set_xlabel('Frequency')
ax.set_title("Top 25 Content Words in Macbeth (Shakespeare)")
plt.tight_layout()
plt.show()
```

---

## Step 7 — Bonus Analysis

```python
# 1. How many unique words appear only once (hapax legomena)?
hapax = [w for w, c in word_counts.items() if c == 1]
print(f"Words appearing only once: {len(hapax):,} ({len(hapax)/len(word_counts)*100:.1f}% of vocabulary)")
```

```python
# 2. How many times do the three witches say 'fair' or 'foul'?
for w in ['fair', 'foul']:
    print(f"'{w}': {word_counts.get(w, 0)} times")
```

```python
# 3. Vocabulary richness: type-to-token ratio
ttr = len(set(words)) / len(words)
print(f"Type-to-token ratio: {ttr:.3f}")
print("(1.0 = every word is unique, 0 = one word repeated infinitely)")
```

```python
# 4. Average word length
avg_length = sum(len(w) for w in words) / len(words)
print(f"Average word length: {avg_length:.2f} characters")
```
