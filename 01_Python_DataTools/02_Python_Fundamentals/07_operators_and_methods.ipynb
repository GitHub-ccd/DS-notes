## Operator Precedence

```python
# PEMDAS: Parentheses, Exponents, Multiplication/Division, Addition/Subtraction
print(2 + 3 * 4)       # 14, not 20
print((2 + 3) * 4)     # 20
print(2 ** 3 ** 2)     # 512 — exponentiation is right-associative (2^(3^2) = 2^9)
print((2 ** 3) ** 2)   # 64

# Augmented assignment operators
x = 10
x += 5   # x = x + 5 → 15
x -= 3   # x = x - 3 → 12
x *= 2   # x = x * 2 → 24
x //= 4  # x = x // 4 → 6
x **= 2  # x = x ** 2 → 36
print(x)

# Bitwise operators (useful for flag manipulation and binary data)
print(5 & 3)   # 1   — AND
print(5 | 3)   # 7   — OR
print(5 ^ 3)   # 6   — XOR
print(~5)      # -6  — NOT
print(5 << 1)  # 10  — left shift (multiply by 2)
print(5 >> 1)  # 2   — right shift (divide by 2)
```

## Essential Built-in Functions

```python
data = [3, 1, 4, 1, 5, 9, 2, 6, 5, 3, 5]

# Numeric aggregations
print(len(data))       # 11
print(sum(data))       # 44
print(min(data))       # 1
print(max(data))       # 9
print(round(44/11, 2)) # 4.0
print(abs(-7))         # 7
print(pow(2, 8))       # 256 — equivalent to 2**8

# Type conversions
print(int('42'))       # 42
print(float('3.14'))   # 3.14
print(str(100))        # '100'
print(list('abc'))     # ['a', 'b', 'c']
print(tuple([1, 2]))   # (1, 2)
print(set([1, 1, 2]))  # {1, 2}
print(dict(a=1, b=2))  # {'a': 1, 'b': 2}
```

```python
# any() and all()
scores = [82, 91, 78, 65, 88]
print(all(s >= 60 for s in scores))   # True — everyone passed
print(any(s >= 90 for s in scores))   # True — at least one got ≥ 90

flags = [True, False, True]
print(any(flags))   # True
print(all(flags))   # False

# sorted() vs .sort()
nums = [3, 1, 4, 1, 5]
print(sorted(nums))            # [1, 1, 3, 4, 5] — new list, original unchanged
print(sorted(nums, reverse=True))  # [5, 4, 3, 1, 1]
nums.sort()                    # modifies in place, returns None

# enumerate and zip (recap)
for i, val in enumerate(['a', 'b', 'c'], start=1):
    print(f"{i}: {val}")

for x, y in zip([1, 2, 3], [4, 5, 6]):
    print(x + y)
```

```python
# Introspection
x = [1, 2, 3]
print(type(x))          # <class 'list'>
print(isinstance(x, list))  # True
print(dir(x))           # list all attributes and methods
help(x.append)          # show docstring — useful in REPL
```

## String Methods (Data Cleaning Reference)

```python
# Quick reference — most useful for data wrangling
raw = "  HELLO, World!  "

print(raw.strip())           # Remove surrounding whitespace
print(raw.lower())           # Lowercase
print(raw.upper())           # Uppercase
print(raw.title())           # Title Case
print(raw.replace('Hello', 'Hi', 1))  # Replace (max 1 occurrence)
print(raw.split(','))        # Split on comma
print(raw.strip().count('l'))  # Count occurrences
print(raw.strip().find('World'))  # Index of first match
print(raw.strip().startswith('H'))  # Boolean
print(raw.strip().endswith('!'))    # Boolean
print(raw.strip().isdigit())        # All digits?
print(raw.strip().isalpha())        # All letters?
print(' '.join(['join', 'these', 'words']))  # Join list → string
```

## User Input and Output

```python
# print() — flexible output
print('a', 'b', 'c')           # 'a b c' — default sep=' '
print('a', 'b', 'c', sep=', ') # 'a, b, c'
print('no newline', end='')    # suppress newline
print()  # blank line

# input() — always returns a string, need to convert for numbers
# (In notebooks, this works interactively)
# name = input("Your name: ")
# age = int(input("Your age: "))

# Simulated version for notebook:
simulated_age = "25"
age = int(simulated_age)
print(f"In 10 years you'll be {age + 10}")
```

```python
# File I/O
# Writing
with open('temp_output.txt', 'w') as f:
    f.write('Line 1\n')
    f.write('Line 2\n')
    print('Line 3', file=f)  # print to file

# Reading
with open('temp_output.txt', 'r') as f:
    content = f.read()
    print(content)

# Reading line by line
with open('temp_output.txt', 'r') as f:
    for line in f:
        print(line.rstrip())  # strip newline

# The 'with' statement is important — it ensures the file is closed even if an error occurs

import os
os.remove('temp_output.txt')  # cleanup
```
