# Classes and Object-Oriented Programming

## Introduction

Python is an object-oriented language — nearly everything in it is an object. The `int`, `str`, and `list` you've been using are all classes. OOP lets you define your own types: bundling related data and behavior into a single reusable unit called a **class**.

## Why OOP?

Early programs were **procedural** — a sequence of steps with global variables. As programs grew, this became unmanageable: any piece of code could change any variable at any time.

**Functions** solved the reuse problem. **Objects** solved the organisation problem — they bundle the *data* (attributes) and the *functions that act on that data* (methods) into a single named unit.

You've already been using OOP without realising it:

```python
s = "hello"          # s is an instance of the str class
s.upper()            # .upper() is an instance method

df = pd.DataFrame()  # df is an instance of the DataFrame class
df.describe()        # .describe() is an instance method
```

## Classes and Instances

````python
# Define a class with the 'class' keyword
class Dog:
    pass   # empty class for now

# Instantiate — create an instance object from the class blueprint
rex  = Dog()
fido = Dog()

print(type(rex))          # <class '__main__.Dog'>
print(rex is fido)        # False — two separate objects
print(isinstance(rex, Dog))  # True
````

The **class** is the blueprint. The **instance** is the actual object built from that blueprint. You can create as many instances as you like.

---

## Instance Methods and `self`

````python
class Dog:

    def bark(self):
        return 'Woof!'

    def who_am_i(self):
        return self   # returns the instance itself

rex = Dog()
print(rex.bark())         # 'Woof!'
print(rex.who_am_i())     # <__main__.Dog object at 0x...>
print(rex == rex.who_am_i())  # True — self IS the instance
````

### Why does every method need `self`?

When you call `rex.bark()`, Python translates this to `Dog.bark(rex)` — the instance is **implicitly passed as the first argument**. `self` is just the conventional name for that argument. Without it:

```python
class Dog:
    def bark():          # Missing self
        return 'Woof!'

Dog().bark()             # TypeError: bark() takes 0 arguments but 1 was given
```

## `__init__` — Initialising Instance Attributes

````python
class Dog:

    def __init__(self, name, breed, age):
        # Instance attributes — unique to each object
        self.name  = name
        self.breed = breed
        self.age   = age

    def bark(self):
        return f'{self.name} says: Woof!'

    def describe(self):
        return f'{self.name} is a {self.age}-year-old {self.breed}.'


rex  = Dog('Rex',  'German Shepherd', 4)
fido = Dog('Fido', 'Labrador',        2)

print(rex.bark())        # Rex says: Woof!
print(fido.describe())   # Fido is a 2-year-old Labrador.

# Each instance has its own copy of the attributes
print(rex.name)          # Rex
print(fido.name)         # Fido
````

`__init__` is called automatically when you write `Dog(...)`. It's not the constructor itself — Python creates the object first — but it's where you set up the initial state.

---

## Class Attributes — Shared Across All Instances

````python
class Dog:
    # Class attribute — defined on the class, shared by every instance
    species = 'Canis lupus familiaris'

    def __init__(self, name, age):
        self.name = name   # instance attribute — unique per object
        self.age  = age

rex  = Dog('Rex', 4)
fido = Dog('Fido', 2)

print(rex.species)         # Canis lupus familiaris
print(fido.species)        # Canis lupus familiaris  — same value
print(Dog.species)         # accessible on the class directly too

# Instance attributes shadow class attributes for that instance only
rex.species = 'Mystery wolf'
print(rex.species)         # Mystery wolf  (instance attribute now shadows class)
print(fido.species)        # Canis lupus familiaris  (unchanged)
````

## A More Realistic Example — A Data Container

````python
class Student:

    def __init__(self, name, scores):
        self.name   = name
        self.scores = scores  # list of numeric scores

    def average(self):
        return sum(self.scores) / len(self.scores)

    def grade(self):
        avg = self.average()
        if avg >= 90:   return 'A'
        elif avg >= 80: return 'B'
        elif avg >= 70: return 'C'
        elif avg >= 60: return 'D'
        else:           return 'F'

    def add_score(self, score):
        self.scores.append(score)

    def __repr__(self):
        # __repr__ controls what print() shows
        return f'Student({self.name!r}, avg={self.average():.1f}, grade={self.grade()})'


alice = Student('Alice', [88, 92, 79, 95])
bob   = Student('Bob',   [72, 68, 75, 70])

print(alice)            # Student('Alice', avg=88.5, grade=B)
print(bob)

alice.add_score(100)
print(f"After extra credit: {alice}")  # avg updates automatically
````

## The Connection to Libraries You Already Use

Every Pandas and Scikit-learn object is a class instance. When you write:

```python
df = pd.DataFrame({'a': [1, 2, 3]})
df.shape       # instance attribute
df.describe()  # instance method
```

...you're using exactly the same OOP mechanics as above. Scikit-learn models follow an especially clean pattern:

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()      # instantiate
model.fit(X_train, y_train)     # instance method — stores learned params as self.coef_
model.predict(X_test)           # instance method — uses self.coef_ to compute output
model.coef_                     # instance attribute set by .fit()
```

Every estimator in Scikit-learn is a class with at minimum `fit()` and `predict()` methods. When you reach Module 03, you'll see this pattern everywhere.

## Summary

| Concept | Syntax | Purpose |
|---------|--------|---------|
| Define a class | `class Name:` | Blueprint for objects |
| Instantiate | `obj = Name(args)` | Create an instance |
| `__init__` | `def __init__(self, ...)` | Set up initial state |
| Instance attribute | `self.x = value` | Data belonging to one object |
| Class attribute | `x = value` (in class body) | Data shared by all instances |
| Instance method | `def method(self):` | Behaviour belonging to an object |
| `self` | First parameter of every method | Reference to the calling instance |
| `__repr__` | `def __repr__(self):` | Controls `print()` / string output |

OOP is not just a syntax exercise — the entire data science ecosystem (Pandas, Scikit-learn, PyTorch) is built around classes. Understanding this foundation makes reading library source code and writing custom components much easier.
