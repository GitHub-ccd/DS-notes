> **2026 context:** SVMs are less dominant in practice than they were pre-2015 — gradient boosted trees and neural networks outperform them on most tabular and image tasks. They remain useful for small-to-medium datasets with clear margin separation and in high-dimensional spaces (e.g. text classification). SHAP does not have a native fast explainer for SVMs; for explainability on SVM predictions use `shap.KernelExplainer` (model-agnostic, slower) — see [05_shap_explainability.ipynb](../12_Pipelines/S32_05_shap_explainability.ipynb).
