> **2026 context:** XGBoost remains widely used, but **LightGBM** (`lightgbm.LGBMClassifier`) is now often preferred for tabular data — it trains faster (leaf-wise growth vs level-wise), handles large datasets with lower memory, and frequently matches or beats XGBoost accuracy with less tuning. **CatBoost** is a strong alternative when the data has high-cardinality categoricals. For most new tabular ML projects, benchmark LightGBM first. See the [LightGBM docs](https://lightgbm.readthedocs.io/) for a direct sklearn-compatible API.
>
> For model explanations across any of these boosting libraries, SHAP's `TreeExplainer` is the standard tool — see [S32_05_shap_explainability.ipynb](../12_Pipelines/S32_05_shap_explainability.ipynb).
