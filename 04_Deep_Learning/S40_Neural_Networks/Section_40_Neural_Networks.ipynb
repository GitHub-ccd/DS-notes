> **2026 context:** This section uses **Keras** (originally standalone, now `tf.keras` inside TensorFlow). Keras remains a solid high-level API for learning neural networks. However, **PyTorch** (`torch`) has become the dominant framework in research and is now equally common in production — most new papers, HuggingFace models, and ML engineering roles use PyTorch. Key differences:
>
> | | Keras / TensorFlow | PyTorch |
> |---|---|---|
> | Style | High-level, declarative | Lower-level, imperative (Pythonic) |
> | Debugging | Harder (graph execution) | Easier (eager by default) |
> | Research adoption | Declining | Dominant |
> | HuggingFace | Supported | Primary backend |
>
> **Recommendation:** Learn the concepts here with Keras, then transfer to PyTorch when working with HuggingFace models or reading research code. `pip install torch torchvision`
