> **2026 context:** ARIMA remains a solid baseline and is still widely used. Two modern alternatives worth knowing:
>
> - **Prophet** (`prophet` / `neuralprophet`) — Meta's open-source library; fits trend + seasonality components automatically, handles missing data and holidays, minimal tuning. Good for business time series with clear seasonal patterns. `pip install prophet`
> - **NeuralProphet** — extends Prophet with AR-Net and lagged regressors using PyTorch. Better for multi-step forecasting.
> - **Statsforecast** (`statsforecast`) — Nixtla's library; fast ARIMA + ETS + Theta implementations, often outperforms Prophet on benchmark datasets.
>
> For large-scale or multivariate forecasting, **TimeGPT** (Nixtla) and **Chronos** (Amazon) are foundation models for zero-shot time series prediction.
