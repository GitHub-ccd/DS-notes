# Section 39 — Natural Language Processing

> ⚠️ **LEGACY NOTE (2026):** The core content here covers NLTK-based NLP — tokenisation, POS tagging, TF-IDF, bag-of-words, context-free grammars. These are foundational concepts worth understanding, but **NLTK is no longer the primary production NLP tool**. The field has moved to transformer-based models (BERT, RoBERTa, GPT variants) via the HuggingFace ecosystem. **Use instead:** `transformers`, `spaCy` (for production NLP pipelines), `sentence-transformers` (for embeddings). See [14_transformers_huggingface.ipynb](./14_transformers_huggingface.ipynb) for a 2026 starting point.
