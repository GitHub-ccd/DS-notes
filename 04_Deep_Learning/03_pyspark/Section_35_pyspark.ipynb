> **2026 context:** PySpark on a local Docker container is fine for learning, but in production Spark now runs almost exclusively on managed cloud platforms: **Databricks** (available on AWS, Azure, and GCP — the most popular Spark platform), **AWS EMR**, **Azure HDInsight/Synapse**, or **GCP Dataproc**. Databricks adds the Delta Lake format, Unity Catalog, and MLflow integration on top of Spark — making it the de-facto standard for large-scale ML pipelines. If you're working with Spark in a professional context, expect to use Databricks rather than a self-managed cluster.
