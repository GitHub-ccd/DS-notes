> **2026 context:** SVD and ALS-based collaborative filtering are still widely used, but LLM-era recommendation systems increasingly use **dense embeddings** instead of sparse user-item matrices. The modern pattern:
> 1. Encode items (products, articles, songs) as embedding vectors using a transformer model or fine-tuned LLM
> 2. Store embeddings in a **vector database** (Pinecone, Weaviate, ChromaDB, pgvector)
> 3. At query time, embed the user's context/history and retrieve nearest neighbours by cosine similarity
>
> This approach handles cold-start better, naturally incorporates text/metadata, and scales to billions of items. Libraries: `sentence-transformers`, `faiss`, `hnswlib`. See MOD_5 (RAG & Retrieval) for the underlying retrieval mechanics.
